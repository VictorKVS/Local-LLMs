****Домашнее задание**:**



Исследование влияния параметров генерации на поведение LLM

1. Выберите одну любую локальную LLM модель.  
2. Используйте два вопроса:
   - Вопрос 1 (обязательный):  
    "Напиши короткий рассказ (3–4 предложения) о том, как ты нашел загадочную записку в неожиданном месте."
     Этот вопрос чувствителен к креативности, длине и стилю, что позволит увидеть эффект от изменения параметров.
   - Вопрос 2 (придумайте сами):  
     Придумайте свой "ловушечный" или творческий вопрос, на котором можно наблюдать влияние параметров.

3. Проведите три независимых эксперимента (по одному на каждый параметр):  

  Эксперимент 1: `temperature`
   Зафиксируйте: `top_p=0.9`, `max_new_tokens=128`
   - Протестируйте модель с тремя значениями `temperature`: с минимальным, средним и завышеным.

  Эксперимент 2: `top_p`
   - Зафиксируйте: `temperature=0.7`, `max_new_tokens=128`
   - Протестируйте модель с:
     - `top_p=0.1` (узкий выбор токенов)
     - `top_p=0.8` (по умолчанию)
     - `top_p=0.99` (широкий выбор)

 Эксперимент 3: `max_new_tokens`
   - Зафиксируйте: `temperature=0.7`, `top_p=0.9`
   - Протестируйте модель с:
     - `max_new_tokens=64` (очень короткий ответ)
     - `max_new_tokens=128` (по умолчанию)
     - `max_new_tokens=512` (длинный, развёрнутый ответ)

4. Для каждого эксперимента сохраните все ответы

5. На основе проведённых экспериментов с моделью составьте отчёт, отвечая на вопросы:

- **Как изменение `temperature` влияет на креативность, связность и завершённость рассказа (Вопрос 1)?**  
   Сравните ответы при `temperature=0.1`, `0.7` и `1.2`. В каком случае рассказ наиболее соответствует требованию «3–4 предложения»?

- **Как параметры генерации (`temperature`, `top_p`, `max_new_tokens`) влияют на логическую корректность и стабильность ответа на Вопрос 2?**  
   Удалось ли модели дать правильный ответ? Как модель ведёт себя при разных настройках — уверенно ошибается, уходит от ответа или пытается рассуждать?

- **В каком эксперименте и при каких параметрах модель даёт наиболее подходящий ответ по стилю, содержанию и логике?**  
   Обоснуйте выбор: почему именно эта комбинация параметров оказалась оптимальной?

- **Наблюдались ли негативные эффекты при экстремальных значениях параметров?**  
   Например:  
   - бессвязность или странные образы при высоком `temperature`,  
   - обрыв смысла при малом `max_new_tokens`,  
   - повторяющиеся или шаблонные фразы при низком `top_p`.

 **Какой из трёх параметров — `temperature`, `top_p` или `max_new_tokens` — оказал наибольшее влияние на качество генерации?**  
   Аргументируйте: какой параметр сильнее всего меняет характер ответа — и почему его стоит настраивать в первую очередь.
