## **Облако или локальный контур: в чём разница и когда что использовать**



LLM могут развертываться двумя принципиально разными способами: в облаке — на удалённых серверах провайдера, и локально — на собственных аппаратно-программных ресурсах пользователя. Выбор способа развёртывания определяет где хранятся данные и какие возможности настройки доступны.

---

### **Что значит «LLM в облаке»**

Когда модель работает в облаке, она размещена на удалённых серверах компании-провайдера — например, OpenAI, Google или Anthropic. Пользователь взаимодействует с ней через веб-интерфейс или API.

Примеры:
- Использование ChatGPT через сайт chat.openai.com
- Вызов API OpenAI из вашего приложения

Преимущества:

- Высокая скорость ответа — мощные серверы провайдера.
- Не нужно заботиться о техническом обслуживании.

Недостатки:
- Данные передаются на серверы провайдера.
- Оплата за запросы или лимит бесплатного доступа.
- Требуется интернет-соединение.

---

### **Что значит «LLM в локальном контуре»**

Локальный запуск означает, что модель устанавливается и работает непосредственно на вашем оборудовании — компьютере или сервере. Вы полностью управляете процессом: от загрузки модели до формата запросов и обработки данных.Такие модели также называют on-device, offline, self-hosted или private LLM — в зависимости от контекста.

Примеры:
- Установка LM Studio на ноутбук и загрузка туда модели Mistral.
- Запуск модели с Hugging Face.

Преимущества:
- Данные остаются в вашем распоряжении — не передаются третьим лицам.
- Полный контроль над настройкой: промпты, RAG, дообучение, базы знаний.
- Работа без интернета после загрузки весов.
- Отсутствие платы за использование после настройки (только затраты на железо/электричество).

Недостатки:
- Требуется время на установку и первоначальную настройку.
- Нужны минимальные технические навыки.
- Может потребоваться мощное оборудование.



# **Обзор open-source моделей с поддержкой русского языка**

При выборе модели для локального запуска важно учитывать не только её размер и производительность, но и **качество работы с русским языком**. ## **1. Qwen/Qwen1.5-7B-Chat**

- **Параметры**: 7B  
- **Лицензия**: Apache 2.0 — разрешено коммерческое использование без ограничений  
- **Языки**: **русский**, английский, китайский, испанский, французский, немецкий и др. (официально поддерживается более 30 языков)

- **Специализация**: мультиязычные задачи, RAG, генерация кода, длинные контексты (до 32K токенов)  
- **Особенности**: основана на архитектуре Transformer с SwiGLU, QKV bias и group query attention. Улучшенный токенизатор для кода и естественных языков. Прошла пост-обучение с SFT и DPO.  


>  Рекомендуемая версия: `Qwen/Qwen1.5-7B-Chat`

# **Как выбрать выбрать подходящую модель?**


 [Список наиболее подходящих моделей для работы с текстомe](https://huggingface.co/models?pipeline_tag=text-generation&language=ru&sort=trending), можно отфильтровать также по языку.

При работе с локальными моделями важно учитывать не только доступные вычислительные ресурсы, но и то, насколько хорошо модель справляется с  конкретными задачами — будь то следование инструкциям, решение математических задач, генерация кода или ведение диалога.

 **[Open LLM Leaderboard](https://huggingface.co/datasets/open-llm-leaderboard/contents)** от Hugging Face. Этот лидерборд автоматически оценивает тысячи моделей по единым бенчмаркам, позволяя объективно сравнивать их производительность, лицензирование, архитектуру и другие ключевые характеристики.

В таблице ниже приведены основные параметры и краткое пояснение:

##  Параметры Open LLM Leaderboard: что они означают?

| Параметр | Краткое описание |
|---------|------------------|
| **eval_name** | Уникальное имя оценки модели в лидерборде. |
| **Precision** | Точность вычислений (например, `float16`, `bfloat16`). Влияет на скорость, потребление памяти и совместимость с оборудованием. |
| **Type** | Тип модели: `base` (базовая), `chat` (чат-модель), `IFT` (дообучена на инструкциях), `RLHF/DPO` (дообучена с использованием обратной связи) и т.д. |
| **T** | Маркер, указывающий, что модель предназначена для диалога (обычно сопровождается шаблоном общения). |
| **Weight type** | Формат весов модели (часто совпадает с Precision). |
| **Architecture** | Архитектура нейросети: Llama, Mistral, Qwen, Gemma, Yi и др. |
| **Model** | Краткое название модели (например, `Qwen1.5-7B`). |
| **fullname** | Полное имя модели с указанием автора и пути на Hugging Face (например, `01-ai/Yi-1.5-34B-Chat`). |
| **Model sha** | Уникальный хеш коммита модели на Hugging Face Hub — гарантирует воспроизводимость. |
| **Average** | Средний нормализованный балл по всем бенчмаркам — основной показатель общей компетентности модели. |
| **Hub License** | Лицензия модели (MIT, Apache 2.0, commercial и др.). Критически важна для коммерческого использования. |
| **Hub** | Прямая ссылка на страницу модели на Hugging Face. |
| **#Params (B)** | Количество параметров в миллиардах. Чем больше — тем выше потенциал, но и выше требования к RAM/VRAM. |
| **Available on the hub** | Доступна ли модель для скачивания с Hugging Face. |
| **MoE** | Использует ли модель архитектуру «Mixture of Experts» (более эффективна при том же количестве активных параметров). |
| **Flagged** | Помечена ли модель как потенциально ненадёжная (например, из-за некорректных метаданных или аномальных результатов). |
| **Chat Template** | Наличие и тип шаблона форматирования диалога (например, `chatml`, `zephyr`). Обязателен для корректной работы чат-моделей. |
| **CO₂ cost (kg)** | Оценка углеродного следа, связанного с обучением модели. |
| **Upload To Hub Date** | Дата публикации модели на Hugging Face. |
| **Submission Date** | Дата отправки модели на оценку в лидерборд. |
| **Generation** | Настройки генерации (температура, top-p и др.), использованные при тестировании. |
| **Base Model** | Исходная модель, на основе которой создана текущая (актуально для дообученных или merged-моделей). |
| **Official Providers** | Является ли модель официальной от разработчика (Meta, Mistral, Alibaba и др.). |
| **Merged** | Собрана ли модель путём слияния (merge) нескольких других моделей. |

###  Основные бенчмарки (оценочные наборы)

- **IFEval** — способность точно следовать инструкциям.  
- **BBH** (Big-Bench Hard) — сложные задачи логического и абстрактного мышления.  
- **MATH Lvl 5** — математические задачи высокого уровня сложности.  
- **GPQA** — вопросы по естественным наукам на уровне магистратуры.  
- **MUSR** — оценка понимания сценариев и здравого смысла.  
- **MMLU-PRO** — расширенная и усложнённая версия MMLU с защитой от угадывания.

> 💡 при выборе локальной модели ориентируйтесь не только на **Average**, но и на конкретные бенчмарки, соответствующие вашей задаче. Например:  
> - Для программирования — смотрите на модели с высоким **MATH** или специализированные (например, `Yi-Coder`).  
> - Для диалога — важны **IFEval** и наличие корректного **Chat Template**.  
> - Для научных задач — **GPQA** и **BBH**.

##  Как пользоваться таблицей лидерборда

1. **Определите свою задачу**  
   Перед выбором модели чётко сформулируйте, для чего она вам нужна:  
   - Общее общение → смотрите **IFEval**, **Chat Template**, тип `chat`.  
   - Решение задач → обратите внимание на **MATH Lvl 5**, **GPQA**, **BBH**.  
   - Генерация кода → ищите модели с упоминанием `Coder` в названии (например, `Yi-Coder`) или высоким **MATH**.  
   - Работа с длинными текстами → проверьте контекстное окно (часто указано в названии: `-16K`, `-32K`, `-200K`).

2. **Учитывайте ресурсы**  
   - Проверьте **#Params (B)**: модели до 7B часто запускаются на потребительских GPU (например, RTX 3090/4090).  
   - Обратите внимание на **Precision**: `bfloat16` или `float16` работают быстрее и требуют меньше памяти, чем `float32`.  
   - Если вы используете CPU или слабый GPU — ищите **квантованные веса** (обычно не отображаются в лидерборде, но доступны отдельно на Hugging Face).

3. **Проверяйте лицензию**  
   Колонка **Hub License** критически важна:  
   - Для коммерческого проекта избегайте моделей с лицензиями `non-commercial` или `research-only`.  
   - Предпочтительны открытые лицензии: **Apache 2.0**, **MIT**, **Llama Community License** (для моделей Meta).

4. **Обращайте внимание на тип модели**  
   - `base` — не обучена на диалогах, плохо подходит для чата.  
   - `chat`, `IFT`, `DPO`, `RLHF` — оптимизированы для взаимодействия с пользователем.  
   - `fine-tuned on domain-specific datasets` — специализированные модели (медицина, код, финансы и т.д.).

5. **Не гонитесь только за высоким Average**  
   Модель может иметь высокий средний балл, но проваливаться именно в вашей предметной области. Всегда смотрите **сырые результаты по нужным бенчмаркам** (например, `MATH Lvl 5 Raw`).

6. **Проверяйте дату публикации**  
   Новые модели (свежие **Upload To Hub Date**) часто используют улучшенные методы обучения и дают лучшее соотношение «качество/размер».

7. **Используйте фильтрацию на сайте**  
   На [странице лидерборда](https://huggingface.co/datasets/open-llm-leaderboard/contents) можно сортировать по столбцам, искать по названию, фильтровать по архитектуре или количеству параметров.



#  **Инструменты для запуска**



В этом разделе мы рассмотрим три самых популярных решения:

1. **Ollama** — для тех, кто хочет запустить модель одной командой (терминал)
2. **LM Studio** — для тех, кто предпочитает графический интерфейс (GUI)
3. **Hugging Face** — для тех, кто хочет полный контроль через Python

Каждый инструмент подходит для разных задач и уровней подготовки.



## **1. Ollama**



Ollama — это **десктопное приложение с терминальным интерфейсом**.

Вы устанавливаете его, как обычную программу, а затем вводите в терминале:

```bash
ollama run llama3:8b
```

…и модель загружается, настраивается и запускается — автоматически. Можно сразу начать диалог.



Поддерживает множество популярных моделей: Llama 3, Mistral, Phi-3, Gemma, Qwen и другие — все с предустановленными шаблонами чата.

 ✅ Преимущества

- Очень простой старт — не нужно знать Python
- Автоматическая загрузка и настройка моделей
- Работает оффлайн после загрузки
- Есть веб-интерфейс (через `ollama webui` или отдельные проекты)
- Поддержка macOS, Windows, Linux

 ⚠️ Ограничения

- Меньше контроля над параметрами генерации (температура, top_p и т.д. — настраиваются, но не так гибко)
- Нельзя легко подключить RAG или кастомные плагины без доп. инструментов
- Нет визуального редактора промптов
  




> **Для кого**: новички, пользователи Mac/Linux/Windows, кто хочет быстро запустить модель без кода  
> **Особенность**: работает через терминал, автоматически скачивает и настраивает модели

### 1: Установите Ollama


---


Перейдите на [Официальный сайт](https://ollama.com)  → скачайте версию для вашей ОС → установите как обычную программу.

  **macOS:** `.dmg` → перетащите в Applications  
  **Windows:** `.exe` → запустите установщик  
  **Linux:** `.sh` → запустите в терминале

 После установки автоматически откроется окно встроенного веб-интерфейса (Ollama WebUI или Open WebUI). Можно приступать к диалогу с моделью в чате — при первом запросе запустится процесс загрузки выбранной модели.

###   2: Работа с Ollama через терминал

Хотя веб-интерфейс удобен для новичков, **настоящая сила Ollama — в терминале**. Именно так разработчики и исследователи используют его в реальных проектах.Работает **быстрее** — без графического интерфейса. Можно:

-  **автоматизировать** запуск моделей через скрипты
-  **точно контролировать** параметры генерации
-  **интегрировать** с другими инструментами (Python, bash, RAG-системы)


#### **Пример: Запускаем Mistral 7B**

Откройте **терминал** и введите:

```bash
ollama run mistral
```

  Если модель ещё не скачана — Ollama **автоматически загрузит её** (это займёт 2-5 минут).  
  После загрузки — вы сразу попадёте в **интерактивный чат**.

Задайте вопрос:

```
>>> Какое самое высокое здание в мире?
```

→ Модель ответит прямо в терминале — например:

```
 Самое высокое здание в мире - Бурдж-Халифа (Башня Кhalifa) в Дубае, Объединенные Арабские Эмираты. Его высота составляет 828 метров, и он имеет 163 этажа. Бурдж-Халифа был открыт в январе 2010 года и продолжает оставаться самым высоким зданием в мире по состоянию на апрель 2021 года.
```

#  **Тестирование локальных LLM моделей**



Мы хотим **сравнить поведение разных локальных LLM** на одинаковых, простых, но каверзных вопросах, чтобы оценить:

- Насколько модель понимает **контекст и здравый смысл**.
- Насколько она **устойчива к провокациям** (например, к ложной логике вроде *«2 термометра = 30 градусов»*).
- Как **формулирует ответ** — чётко и логично, с пояснением, или механически, без анализа.


Каждый из вопросов — это «когнитивная ловушка», проверяющая способность модели выйти за рамки буквального прочтения и применить **реальный мир, здравый смысл и контекстное мышление**.




 > 1. **«Термометр показывает +15°C. Сколько покажут два таких термометра?»**

 **Проверяет понимание физических величин**: температура — это интенсивная величина, она **не складывается** при добавлении приборов.  
 Ожидаемый ответ: **«+15°C»** — с пояснением, что показания термометров не суммируются.  
 **Ловушка**: соблазн механически сложить числа (15 + 15 = 30).

 > 2. **«На дереве сидели 7 воробьёв. Одного съел кот. Сколько осталось?»**

 **Проверяет контекстное и причинно-следственное мышление**: воробьи — живые существа, реагирующие на угрозу.  
 Ожидаемый ответ: **«Скорее всего, ни одного — остальные улетели от страха»**.  
 **Ловушка**: формальный подсчёт (7 – 1 = 6), игнорирующий поведение птиц.

 > 3. **«По небу летели голубь, муха и сокол. Сколько птиц летело?»**

 **Проверяет знание категорий и способность к классификации**: муха — **насекомое**, а не птица.  
 Ожидаемый ответ: **«Две птицы — голубь и сокол»**.  
 **Ловушка**: механический подсчёт всех упомянутых существ (3), без анализа их природы.

##  **Описание фреймворка тестирования локальных LLM**


 > **Этап 1: Подготовка тестовых вопросов**

Все модели тестируются на **одинаковом наборе вопросов**, хранящемся в глобальной структуре `TEST_QUESTIONS` → автоматически преобразуется в `TEST_CASES`.

Каждый вопрос — это «когнитивная ловушка», проверяющая:

- **Физику**: температура не складывается.
- **Контекст**: живые существа реагируют на угрозы (воробьи улетят от кота).
- **Классификацию**: муха ≠ птица.
- **Логику и здравый смысл**.



> **Этап 2: Загрузка модели и токенизатора**

Перед тестированием загружаем:

- **Токенизатор** — для преобразования текста в токены и обратно.
- **Модель** — с любыми параметрами (`device_map`, `load_in_8bit`, `torch_dtype` и т.д.).

Фреймворк **не зависит от способа загрузки** — главное, чтобы `model` и `tokenizer` были доступны в глобальной области.

>  **Этап 3: Универсальное тестирование через `test_model_universal`**

Вызываем функцию, которая:

1. Для **каждого вопроса**:
   - Формирует промпт (с `system_prompt`, если нужен — например, для Saiga).
   - Генерирует ответ с замером времени.
   - Декодирует только сгенерированную часть.
   - Логирует промпт и ответ (для отладки).

2. Возвращает **словарь** с:
   - `answers` — список ответов.
   - `times` — список времён генерации.
   - `total_time` — общее время.

> **Этап 4: Добавление результатов в таблицу через `add_to_results`**

Результаты добавляются в **глобальную таблицу `results_df`**.

Функция -  Динамически определяет, какие колонки нужны (`Вопрос 1`, `Ответ 1`, ...).
- Сохраняет:
  - Название модели.
  - Текст вопросов.
  - Ответы.
  - Время генерации.
  - Общее время.
- Оставляет «Итоговый вывод» пустым — для ручного анализа.


> **Этап 5: Сохранение чекпоинта и перезапуск среды**

Для избежания утечек памяти и фрагментации VRAM:

1. После тестирования → вызываем `save_results_checkpoint(results_df, SAVE_PATH)`. Результаты надёжно сохраняются на Google Drive (/content/drive/MyDrive/...) — и не будут потеряны при очистке.
2. **Перезапускаем среду выполнения Colab** `exit()`.
3. Проводим очистку диска → вызываем `clear_content_dir()`. Удаляются все временные файлы, веса моделей, кэши, датасеты (кроме Google Drive)
4. **Повторная инициализация**
- Устанавливаем библиотеки (если нужно).
- Импортируем вспомогательные функции (test_model_universal, add_to_results, и т.д.).
- Инициализируем TEST_CASES, hf_token, конфиги.
- Всё "с нуля" — как в начале ноутбука.


> **Этап 6: Ручной разбор результата тестирования и фиксация итогового вывода**

На этом этапе необходимо провести разбор полученных результатов тестирования и записать их в виде строки:
`final_conclusion = """ указываем итоговые выводы """`

Затем обновляем таблицу через `add_final_conclusion(model_name, conclusion)` — **специальную функцию**, которая корректно заполняет колонку «Итоговый вывод».


> **Этап 7: Финальная визуализация и экспорт**

Вызываем финальный блок кода, который:

- Создаёт сводную таблицу `final_df` — только ключевые колонки.
- **Корректно обрабатывает `NaN`** — показывает пустые строки вместо `'nan'`.
- Обрезает длинные тексты для читаемости.
- Красиво выводит таблицу с форматированием.
- Сохраняет **два CSV-файла**:
  - `llm_comparison_full_results.csv` — полные данные.
  - `llm_comparison_summary.csv` — сводка для презентаций.
- **Сохраняет файлы на Google Drive** — чтобы не потерять после перезапуска.

````python
from openai import OpenAI
import time

client = OpenAI()

# Наши вопросы
TEST_QUESTIONS = [
    "Термометр показывает +15°C. Сколько покажут два таких термометра?",
    "На дереве сидели 7 воробьёв. Одного съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

# Три разных seed для лёгкой стохастичности (тест «качания» ответов)
seeds = [1, 2, 3]

# Гиперпараметры
MODEL = "gpt-4o-mini"
TEMPERATURE = 0.4
MAX_TOKENS = 64

for q in TEST_QUESTIONS:
    answers = []
    for s in seeds:
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=TEMPERATURE,
                max_tokens=MAX_TOKENS,
                seed=s,
                messages=[
                    {"role": "user", "content": q}
                ],
            )
            text = (resp.choices[0].message.content or "").strip()
            answers.append(text)
            time.sleep(0.2)  # пауза между запросами
        except Exception as e:
            # В случае ошибки всё равно печатаем маркер, чтобы не ломать блок
            answers.append(f"[ошибка запроса: {e}]")

    # Печатаем ТОЛЬКО ответы модели (3 строки), затем пустую строку-разделитель
    print("\n".join(answers))
    print()
````

На дереве сидели 7 воробьёв. Одного съел кот. Сколько осталось?

Ответы модели:
* На дереве осталось 6 воробьёв. Однако, если учитывать, что воробей, которого съел кот, больше не на дереве, то можно сказать, что на дереве остались 6 воробьёв, но если говорить о ситуации в целом, то, возможно, воробьи
* На дереве осталось 6 воробьёв. Однако, если учитывать, что воробьи могут испугаться и улететь, то можно сказать, что на дереве не осталось ни одного воробья. Но в строгом смысле, если считать только тех, кто остался на дереве,
* На дереве осталось 6 воробьёв. Однако, если учитывать, что воробей был съеден котом, то можно сказать, что на дереве всё равно сидят 7 воробьёв, просто один из них теперь в животе кота.

Оцените логику ...

Оценка логики:

1. «Осталось 6… (обрыв)»
   — Формально-арифметически верно (7−1=6). Но ответ растекается и обрывается на полуслове. Недостатки: избыточные оговорки, нет чёткого финала.

2. «Осталось 6… но если испугались — 0… (строго — 6)»
   — Логика в порядке: корректно различает два контекста — «строгий счёт» (6) vs. «жизненная ситуация» (0). Недостатки: многословие и отсутствие окончательного, единственного вывода.

3. «…всё равно сидят 7, один в животе кота»
   — Нелогично и фактически неверно. Птица, съеденная котом, не «сидит на дереве». Минус балл за здравый смысл.

Что считать «правильным»?

* Формальный ответ: **6**.
* С учётом поведения птиц (реалистичный сценарий): **0** (испугались и улетели).
  Лучший ответ даёт число и контекст, например: «**6** (если считать формально); **0**, если учесть, что остальные, вероятно, улетели».

Как улучшить стабильность и качество:

* Понизить `temperature` до 0.0–0.2.
* Дать строгую инструкцию:
  *Системное сообщение*: «Отвечай: 1) короткое число-ответ; 2) одно предложение пояснения. Без лишних оговорок.»
* Пример целевого формата:
  «**6** — формально 7−1=6; в реальности могли улететь, но просили формальный счёт».


````python


from openai import OpenAI
import time

client = OpenAI()

seeds = [1, 2, 3]

MODEL = "gpt-4o-mini"
TEMPERATURE = 0.1
MAX_TOKENS = 64

# Наши вопросы
TEST_QUESTIONS = [
    "Термометр показывает +15°C. Сколько покажут два таких термометра?",
    "На дереве сидели 7 воробьёв. Одного съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

SYSTEM_PROMPT = (
    "Отвечай строго в формате: 1) короткое число-ответ; 2) одно короткое "
    "предложение пояснения. Никаких оговорок и альтернативных трактовок."
)

for q in TEST_QUESTIONS:
    for s in seeds:
        try:
            resp = client.chat.completions.create(
                model=MODEL,
                temperature=TEMPERATURE,
                max_tokens=MAX_TOKENS,
                top_p=1.0,
                seed=s,
                messages=[
                    {"role": "system", "content": SYSTEM_PROMPT},
                    {"role": "user", "content": q},
                ],
            )
            text = (resp.choices[0].message.content or "").strip()
            print(text)
            time.sleep(0.2)
        except Exception as e:
            print(f"[ошибка запроса: {e}]")
    print()  # пустая строка-разделитель между вопросами
````

##  **Вспомогательные функции для автоматизации тестов**

````python
# ===========================================
#  ГЛОБАЛЬНЫЕ СТРУКТУРЫ И ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ ДЛЯ ТЕСТИРОВАНИЯ МОДЕЛЕЙ
# ===========================================

# --------------------------------------------------
#  1. Фиксированные тестовые вопросы
# --------------------------------------------------

TEST_QUESTIONS = [
    "Термометр показывает плюс 15 градусов. Сколько градусов покажут два таких термометра?",
    "На дереве сидели 7 воробьёв, одного из них съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

# --------------------------------------------------
#  2. Автоматически формируем messages для каждого вопроса
# --------------------------------------------------

TEST_MESSAGES = [
    [{"role": "user", "content": q}] for q in TEST_QUESTIONS
]

# --------------------------------------------------
#  3. Глобальный список вопросов для цикла в функциях тестирования
# Структура: [(номер, messages, текст_вопроса), ...]
# → Номера генерируются автоматически
# --------------------------------------------------

TEST_CASES = [
    (i + 1, TEST_MESSAGES[i], TEST_QUESTIONS[i])
    for i in range(len(TEST_QUESTIONS))
]

print(f"✅ Загружено {len(TEST_CASES)} тестовых вопросов.")


# --------------------------------------------------
#  4. Универсальный системный промпт для всех моделей
#  Используется в test_model_universal() для  сравнения
#  Английский язык выбран потому, что:
#    - Все Instruct-модели (Saiga, Mistral, Qwen, DeepSeek, OpenChat) обучены на инструкциях на английском.
#    - Не влияет на язык ответа — модель сама отвечает на языке вопроса.
#    - Гарантирует одинаковые условия тестирования для всех моделей.
# --------------------------------------------------

SYSTEM_PROMPT = "You are a helpful, precise, and concise assistant."

print(f"✅ Системный промпт установлен: \"{SYSTEM_PROMPT}\"")


# --------------------------------------------------
#  5. Динамическая инициализация таблицы результатов
#  Автоматически создаёт колонки под любое количество вопросов
# --------------------------------------------------

def init_results_table():
    """
    Инициализирует или возвращает существующую таблицу результатов.
    Колонки динамически формируются под количество вопросов в TEST_CASES.
    """
    global results_df

    # Если таблица уже существует — не пересоздаём
    if 'results_df' in globals() and isinstance(results_df, pd.DataFrame) and not results_df.empty:
        print("ℹ️ Таблица результатов уже существует — оставляем как есть.")
        return results_df

    # Формируем названия колонок динамически — БЕЗ обрезки текста вопроса
    columns = ['Модель']

    for i, (_, _, question_text) in enumerate(TEST_CASES, 1):
        columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])

    columns.append('Общее время (сек)')
    columns.append('Итоговый вывод')  # ← для ручного заполнения после анализа

    # Создаём пустую таблицу
    results_df = pd.DataFrame(columns=columns)
    print(f"✅ Таблица результатов инициализирована под {len(TEST_CASES)} вопросов.")
    return results_df


# Инициализируем таблицу
results_df = init_results_table()


# --------------------------------------------------
# 6. УНИВЕРСАЛЬНАЯ функция добавления результатов в таблицу
#
# --------------------------------------------------

def add_to_results(model_name, answers, times, total_time):
    """
    УНИВЕРСАЛЬНАЯ функция: добавляет результаты тестирования модели в таблицу.
    Поддерживает ЛЮБОЕ количество вопросов.
    Итоговый вывод остаётся пустым — его заполняет пользователь вручную после анализа.

    Параметры:
        model_name : str
            Название модели (например, "Qwen/Qwen1.5-7B-Chat").
        answers : list[str]
            Список сгенерированных ответов — по одному на каждый вопрос.
        times : list[float]
            Список времён генерации — по одному на каждый ответ.
        total_time : float
            Общее время выполнения всех запросов.

    Пример вызова:
        result = test_model_universal(tokenizer, model)
        add_to_results("Saiga-7B", result["answers"], result["times"], result["total_time"])
    """
    global results_df

    # Проверка соответствия длины списков
    if len(answers) != len(times):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а времён {len(times)}")
    if len(answers) != len(TEST_QUESTIONS):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а вопросов {len(TEST_QUESTIONS)}")

    # Создаём новую строку результата
    new_row = {'Модель': model_name}

    # Заполняем данные по каждому вопросу
    for i in range(len(answers)):
        q_num = i + 1

        # Ищем имя колонки
        question_col = f"Вопрос {q_num}"
        if question_col not in results_df.columns:
            raise KeyError(f"Не найдена колонка для вопроса {q_num}: '{question_col}'")

        # Заполняем текст вопроса, ответ и время
        new_row[question_col] = TEST_QUESTIONS[i]
        new_row[f"Ответ {q_num}"] = answers[i]
        new_row[f"Время ответа {q_num} (сек)"] = round(times[i], 2)

    # Заполняем общее время и оставляем итог пустым
    new_row['Общее время (сек)'] = round(total_time, 2)
    new_row['Итоговый вывод'] = ""

    # Добавляем строку в DataFrame
    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)
    print(f"✅ Результаты для {model_name} добавлены в таблицу. Итоговый вывод — заполните вручную.")

# --------------------------------------------------
# 7. Функция добавления итогового вывода по модели
# --------------------------------------------------

def add_final_conclusion(model_name, conclusion):
    """
    Ручное добавление итогового анализа для указанной модели.
    Обновляет существующую строку в таблице results_df.

     Параметры:
        model_name : str
            Название модели, как оно было добавлено ранее (например, "Qwen/Qwen1.5-7B-Chat").
        conclusion : str
            Текст итогового вывода — анализ логики, ошибок, скорости и т.д.

    """
    global results_df

    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов не инициализирована.")
        return

    # Находим индекс строки с указанной моделью
    mask = results_df['Модель'] == model_name
    if not mask.any():
        print(f"❌ Модель '{model_name}' не найдена в таблице.")
        return

    # Обновляем значение в колонке "Итоговый вывод"
    results_df.loc[mask, 'Итоговый вывод'] = conclusion
    print(f"✅ Итоговый вывод для модели '{model_name}' успешно добавлен.")


# --------------------------------------------------
#  8. Функция вывода результатов (улучшенная для сравнения моделей)
# --------------------------------------------------

def show_results():
    """
    Выводит таблицу сравнения моделей с полными ответами и удобным форматированием.
    """
    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов пуста или не инициализирована.")
        return

    print("\n📊 СРАВНЕНИЕ МОДЕЛЕЙ НА ТЕСТОВЫХ ВОПРОСАХ\n")

    # Настройка pandas для полного отображения текста
    pd.set_option('display.max_colwidth', None)
    pd.set_option('display.width', None)
    pd.set_option('display.max_columns', None)

    # Определяем колонки для отображения
    cols_to_show = ['Модель']
    for i in range(1, len(TEST_QUESTIONS) + 1):
        cols_to_show.extend([f'Вопрос {i}', f'Ответ {i}'])
    cols_to_show.append('Общее время (сек)')
    cols_to_show.append('Итоговый вывод')

    # Проверка наличия колонок
    missing = [col for col in cols_to_show if col not in results_df.columns]
    if missing:
        print(f"⚠️ Отсутствуют колонки: {missing}. Возможно, таблица не инициализирована корректно.")
        display(results_df)
        return

    # Формируем сводную таблицу
    summary_df = results_df[cols_to_show].copy()

    # Стилизация
    styled = summary_df.style.set_properties(**{
        'text-align': 'left',
        'white-space': 'pre-wrap',
        'word-wrap': 'break-word',
        'vertical-align': 'top',
        'font-family': 'Consolas, monospace',
        'font-size': '13px'
    }).set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#f0f0f0'), ('font-weight', 'bold')]},
        {'selector': 'td', 'props': [('border', '1px solid #ddd'), ('padding', '10px')]}
    ]).set_caption("📈 Сравнение локальных LLM на тестах логики и креативности")

    display(styled)
````

##  **Служебные функции для работы с результатами (сохранение / загрузка)**

````python
# --------------------------------------------------
#  1. Путь по умолчанию для сохранения результатов на Google Drive
# --------------------------------------------------

DEFAULT_SAVE_PATH = "/content/drive/MyDrive/llm_benchmark/llm_comparison_results.csv"


# --------------------------------------------------
# 2. загрузка результатов с адаптацией структуры
# Загружает CSV и автоматически приводит к актуальной схеме вопросов
# Если файл отсутствует — создаёт новую таблицу
# Гарантирует совместимость с add_to_results()
# --------------------------------------------------

def load_or_init_results(test_questions=TEST_QUESTIONS, filepath=DEFAULT_SAVE_PATH):
    """
    Загружает сохранённые результаты из CSV и адаптирует структуру под текущие вопросы.
    Если файла нет — создаёт новую таблицу.
    Возвращает DataFrame `results_df`.

    Параметры:
        test_questions : list — список вопросов (по умолчанию TEST_QUESTIONS)
        filepath : str — путь к файлу CSV

    Использование:
        results_df = load_or_init_results()
    """
    #  Формируем ожидаемую структуру колонок на основе текущих вопросов
    expected_columns = ['Модель']
    for i in range(1, len(test_questions) + 1):
        expected_columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])
    expected_columns.extend(['Общее время (сек)', 'Итоговый вывод'])

    # Проверяем наличие файла
    if os.path.exists(filepath):
        try:
            # Загружаем данные
            loaded_df = pd.read_csv(filepath)
            print(f"📂 Загружены предыдущие результаты из: {filepath}")

            #  Проверяем, совпадает ли структура
            if set(loaded_df.columns) == set(expected_columns):
                results_df = loaded_df.copy()
                print(f"✅ Структура актуальна. В таблице {len(results_df)} моделей.")
            else:
                # Адаптируем старую таблицу под новую структуру
                print("⚠️ Структура устарела — создаём обновлённую версию...")
                results_df = pd.DataFrame(columns=expected_columns)

                # Переносим совместимые данные
                for _, row in loaded_df.iterrows():
                    new_row = {}
                    # Общие поля
                    for col in ['Модель', 'Общее время (сек)', 'Итоговый вывод']:
                        new_row[col] = row.get(col, "")
                    # Поля по каждому вопросу
                    for i in range(1, len(test_questions) + 1):
                        for ft in ['Вопрос', 'Ответ']:
                            new_row[f"{ft} {i}"] = row.get(f"{ft} {i}", "")
                        new_row[f"Время ответа {i} (сек)"] = row.get(f"Время ответа {i} (сек)", 0.0)
                    # Добавляем строку
                    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)

                print(f"✅ Создана обновлённая таблица с {len(results_df)} перенесёнными записями.")

        except Exception as e:
            # При ошибке — создаём новую таблицу
            print(f"❌ Ошибка загрузки: {e}. Создаём новую таблицу.")
            results_df = pd.DataFrame(columns=expected_columns)
    else:
        # Файл не найден — инициализируем с нуля
        print("🆕 Файл не найден — создаём новую таблицу.")
        results_df = pd.DataFrame(columns=expected_columns)
        print(f"✅ Таблица инициализирована под {len(test_questions)} вопросов.")

    # Сбрасываем индекс для чистоты
    return results_df.reset_index(drop=True)


# --------------------------------------------------
# 3. Cохранение чекпоинта на Google Drive
# Создаёт папку, если её нет
# Кодировка utf-8-sig для корректного открытия в Excel
# Выводит подтверждение — можно перезапускать среду
# --------------------------------------------------

def save_results_checkpoint(df, filepath=DEFAULT_SAVE_PATH):
    """
    Сохраняет DataFrame в CSV на Google Drive.

    Параметры:
        df : pd.DataFrame — таблица для сохранения
        filepath : str — путь к файлу CSV

    Использование:
        save_results_checkpoint(results_df)
    """
    # Создаём директорию, если не существует
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    # Сохраняем с кодировкой utf-8-sig
    df.to_csv(filepath, index=False, encoding='utf-8-sig')
    print(f"💾 Чекпоинт сохранён: {filepath}")
    print("✅ Теперь можно перезапустить среду.")
````

## **Универсальная функция тестирования модели**

````python
# ============================================
# УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ LLM
# ============================================

# --- ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ (инкапсулируют технические детали) ---

def _infer_device(model):
    """Определяет, на каком устройстве (CPU/GPU) работает модель."""
    try:
        return next(model.parameters()).device
    except StopIteration:
        return torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")

def _ensure_padding_ids(tokenizer, model):
    """Гарантирует, что pad_token_id задан — иначе используем eos_token_id."""
    if tokenizer.pad_token_id is None and tokenizer.eos_token_id is not None:
        tokenizer.pad_token_id = tokenizer.eos_token_id

def _maybe_inject_system(messages, system_prompt):
    """Добавляет системный промпт в начало."""
    if not system_prompt:
        return list(messages)
    if any(m.get("role") == "system" for m in messages):
        return list(messages)
    return [{"role": "system", "content": system_prompt}] + list(messages)

def _build_fallback_prompt(messages, system_prompt=None):
    """Создаёт простой текстовый промпт, если у токенизатора нет chat_template."""
    parts = []
    if system_prompt:
        parts.append(f"system\n{system_prompt}\n")
    for m in messages:
        parts.append(f"{m.get('role', 'user')}\n{m.get('content', '')}\n")
    parts.append("assistant\n")
    return "".join(parts)


# --- ОСНОВНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ ---

def test_model_universal(
    tokenizer,
    model,
    device=None,
    *,
    system_prompt=None,
    generation_config=None,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    show_prompts=True
):
    """
    Тестирует модель на всех вопросах из TEST_CASES.
    Возвращает структурированный словарь с ответами, временами и общим временем.
    Подходит для любых моделей и любого числа вопросов.
    """

    # Подготавливаем окружение
    _ensure_padding_ids(tokenizer, model)
    if device is None:
        device = _infer_device(model)

    answers, times = [], []
    total_time = 0.0

    # Если нет вопросов
    if not TEST_CASES:
        print("⚠️  Нет тестовых вопросов — возвращаем пустой результат.")
        return {"answers": [], "times": [], "total_time": 0.0}

    # Обрабатываем каждый вопрос
    for question_num, messages, question_text in TEST_CASES:
        print(f"🔹 Вопрос {question_num}: {question_text[:80]}...")

        # Формируем сообщения с системным промптом (если нужен)
        msgs = _maybe_inject_system(messages, system_prompt)

        # Генерируем промпт — через шаблон или fallback
        try:
            prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        except (ValueError, AttributeError, NotImplementedError):
            print("⚠️  Без chat_template — используем fallback-формат.")
            prompt = _build_fallback_prompt(msgs, system_prompt=system_prompt)

        if show_prompts:
            print(f"📝 Промпт:\n{prompt}\n{'='*60}")

        # Токенизируем и переносим на устройство
        inputs = tokenizer(prompt, return_tensors="pt", add_special_tokens=False)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        # Генерируем ответ
        start_time = time.time()
        with torch.no_grad():
            if generation_config is not None:
                # Используем переданный конфиг генерации
                gc = generation_config
                if getattr(gc, "pad_token_id", None) is None and tokenizer.pad_token_id is not None:
                    gc = gc.clone()
                    gc.pad_token_id = tokenizer.pad_token_id
                outputs = model.generate(**inputs, generation_config=gc)
            else:
                # Используем стандартные параметры
                outputs = model.generate(
                    **inputs,
                    max_new_tokens=max_new_tokens,
                    do_sample=True,
                    temperature=temperature,
                    top_p=top_p,
                    pad_token_id=tokenizer.pad_token_id,
                    eos_token_id=getattr(model.config, "eos_token_id", None)
                )

        gen_time = time.time() - start_time
        total_time += gen_time

        # Декодируем только сгенерированную часть (после входного промпта)
        out_ids = outputs[0]
        in_len = inputs["input_ids"].shape[1]
        gen_ids = out_ids[in_len:]
        answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

        # Выводим результат
        print(f"✅ Ответ {question_num} за {gen_time:.2f} сек")
        print(f"💬 {answer}\n")

        answers.append(answer)
        times.append(gen_time)

    # Возвращаем всё в структурированном виде — для любого числа вопросов
    return {
        "answers": answers,
        "times": times,
        "total_time": total_time
    }
````

**ОПИСАНИЕ УНИВЕРСАЛЬНОЙ ФУНКЦИИ ТЕСТИРОВАНИЯ LLM**
При тестировании разных моделей важно **соблюдать одинаковые условия**:

- Одинаковые вопросы.
- Одинаковый формат промптов.
- Одинаковые параметры генерации.
- Замер времени и качества ответов.

Функция `test_model_universal` позволяет единообразно тестировать любую LLM на любом наборе вопросов, независимо от её архитектуры (Qwen, Saiga, LLaMA и др.).





---

 **Архитектура функции**

Функция состоит из:

1. **Вспомогательных подфункций** — `_infer_device`, `_ensure_padding_ids`, `_maybe_inject_system`, `_build_fallback_prompt`

 Они решают **технические задачи**, чтобы основная функция оставалась **чистой и понятной**.

>  **Почему они начинаются с `_`?**  
> В Python символ `_` в начале имени — это **соглашение о приватности**.  
> Такие функции предназначены **только для внутреннего использования** внутри модуля.  
> Это не техническое ограничение, а **сигнал разработчику**:  
> *«Не вызывай меня напрямую — я часть внутренней реализации, могу измениться»*.  
> Это стандартная практика в Python (PEP8) и используется в библиотеках вроде `transformers`, `pandas`, `scikit-learn`.

---

2. **Основного цикла по `TEST_CASES`**

 Список вопросов в формате:  
```python
TEST_CASES = [
    (1, [{"role": "user", "content": "Вопрос 1..."}], "Текст вопроса 1"),
    (2, [{"role": "user", "content": "Вопрос 2..."}], "Текст вопроса 2"),
    ...
]
```

---
3. **Генерации промпта**

 Пробует использовать `tokenizer.apply_chat_template` — **официальный и правильный способ** формировать промпты для чат-моделей.  
 Если шаблона нет — использует **резервный формат** вида:

```
system
Ты — полезный ассистент.

user
Какой сегодня день?

assistant
```



4. **Токенизации и генерации** — с замером времени

 Переносит тензоры на нужное устройство (CPU/GPU).  
 Замеряет время генерации.  
 Поддерживает как **кастомный `generation_config`**, так и **стандартные параметры** (`temperature`, `top_p` и т.д.).



5. **Декодирования и возврата результата** — в виде словаря

 **Важно**: отрезает только **сгенерированную часть** — то, что модель добавила после промпта.  
 Это делается по длине входных токенов:  
```python
in_len = inputs["input_ids"].shape[1]
gen_ids = out_ids[in_len:]
```

 Функция возвращает **словарь (dict)** со следующими ключами:

```python
{
    "answers":     [str, str, ...],     # список сгенерированных ответов (по одному на вопрос)
    "times":       [float, float, ...], # список времён генерации для каждого вопроса (в секундах)
    "total_time":  float                # общее время выполнения всех запросов
}
```


 Такой формат **легко расширять, логировать, анализировать**.

---

 **Как она работает:**

1. Подготовка

- Определяет устройство (CPU/GPU) — через `_infer_device(model)`.
- Убеждается, что `pad_token_id` задан — иначе генерация может сломаться (`_ensure_padding_ids`).
- Создаёт копию сообщений с системным промптом, если он нужен (`_maybe_inject_system`).



2. Формирование промпта

- Пробует использовать `tokenizer.apply_chat_template(...)` — **рекомендуемый способ**.
- Если не поддерживается — использует fallback через `_build_fallback_prompt`.



3. Токенизация и генерация

- Кодирует промпт → создаёт тензоры → переносит на устройство.
- Замеряет время до/после `.generate()`.
- Поддерживает:
  - Готовый `generation_config` (если передан).
  - Или стандартные параметры: `max_new_tokens`, `temperature`, `top_p`.



4. Обработка результата

- Декодирует **только сгенерированную часть** (после входного промпта).
- Убирает спецсимволы (`skip_special_tokens=True`).
- Добавляет ответ и время в списки.



5. Возврат

- Возвращает всё в **структурированном словаре** — для любого числа вопросов.

```python
def test_model_universal(
    tokenizer,
    model,
    device=None,
    *,
    system_prompt=None,
    generation_config=None,
    max_new_tokens=200,
    temperature=0.7,

):

Функция `test_model_universal` принимает как **обязательные**, так и **опциональные параметры**.
Ниже — подробное описание каждого из них, с пояснениями, зачем он нужен и как влияет на поведение модели.

---

 ➤`tokenizer` *(обязательный параметр)*  
**Тип:** `PreTrainedTokenizer` (например, `AutoTokenizer.from_pretrained(...)`)  
Это токенизатор, **соответствующий загруженной модели**. Он отвечает за:
- Преобразование текстового промпта в последовательность токенов (кодирование).
- Преобразование сгенерированных моделью токенов обратно в читаемый текст (декодирование).  
**Важно**: токенизатор и модель должны быть совместимы — иначе результаты будут некорректны или возникнет ошибка.

---

➤ `model` *(обязательный параметр)*  
**Тип:** `PreTrainedModel` (например, `AutoModelForCausalLM.from_pretrained(...)`)  
Это сама языковая модель, которую вы хотите протестировать.  
→ Она **должна поддерживать метод `.generate()`** — это стандартный интерфейс генерации текста в библиотеке Hugging Face Transformers.  
→ Модель может быть загружена на CPU или GPU — если устройство не указано явно, функция определит его автоматически.

---

➤ `device` *(опциональный параметр)*  
**Тип:** `torch.device`  
**По умолчанию:** `None`  
Задаёт, на каком устройстве выполнять генерацию — например, `"cuda"`, `"cpu"`, или `"cuda:1"` для конкретного GPU.  
→ Если параметр не указан (`None`), функция **автоматически определит устройство** с помощью вспомогательной функции `_infer_device(model)`.  


---

➤ `system_prompt` *(опциональный параметр)*  
**Тип:** `str`  
**По умолчанию:** `None`  
Это **системное сообщение** — инструкция для модели, например:  
> *«Ты — логически мыслящий ассистент. Отвечай кратко и по делу.»*  
→ Если параметр передан, функция **добавит его в начало диалога**, но **только если в исходных `messages` ещё нет системного сообщения**.  
→ Особенно важно для моделей вроде **Saiga**, которые обучены на диалогах с явным системным промптом и могут хуже работать без него.

---

➤ `generation_config` *(опциональный параметр)*  
**Тип:** `GenerationConfig`  
**По умолчанию:** `None`  
Это **готовая конфигурация генерации** от Hugging Face, которая может включать сложные параметры: `repetition_penalty`, `no_repeat_ngram_size`, `length_penalty`, `num_beams` и другие.  
 **Важно**: если вы передаёте `generation_config`, то параметры `max_new_tokens`, `temperature`, `top_p` **игнорируются** — используется только конфиг.  
→ Используйте этот параметр, если вам нужен **полный контроль** над процессом генерации — например, для воспроизводимости экспериментов или тонкой настройки.

---

➤ `max_new_tokens` *(опциональный параметр)*  
**Тип:** `int`  
**По умолчанию:** `200`  
Определяет **максимальное количество новых токенов**, которые модель может сгенерировать за один вызов.  
→ **Не ограничивает длину входного промпта** — только то, что модель добавит после него.  
→ Влияет на:
- Длину ответа (чем больше — тем длиннее может быть ответ).
- Время генерации (чем больше — тем дольше).
- Риск “зацикливания” (если модель генерирует слишком долго).  


---

➤ `temperature` *(опциональный параметр)*  
**Тип:** `float`  
**По умолчанию:** `0.7`  
Параметр **“температуры”** — контролирует **степень случайности** в генерации:  
- `0.0–0.5` → более **детерминированные, предсказуемые** ответы (подходит для логических задач, фактов).  
- `0.7–1.0` → более **креативные, разнообразные** ответы (подходит для творческих задач, диалогов).  
- `>1.0` → высокий риск **бессмыслицы или “галлюцинаций”**.  
💡 Для тестов на логику и здравый смысл — лучше использовать `<0.5`.

---

➤ `top_p` *(опциональный параметр)*  
**Тип:** `float`  
**По умолчанию:** `0.9`  
Параметр **ядерной выборки (nucleus sampling)**.  
→ Модель выбирает следующий токен **только из тех, чья суммарная вероятность ≤ top_p**.  
- `0.9–1.0` → хороший баланс между **разнообразием и качеством**.  
- `0.5` → более **строгий отбор**, ответы становятся более предсказуемыми.  

Часто используется вместе с `temperature` — например, `temperature=0.7, top_p=0.9`.

---

 ➤ `show_prompts` *(опциональный параметр)*  
**Тип:** `bool`  
**По умолчанию:** `True`  
Определяет, **выводить ли сформированный промпт** перед генерацией.  
- `True` → промпт печатается — **очень полезно для отладки**, чтобы убедиться, что он сформирован правильно.  
- `False` → промпт **не выводится** — удобно при массовом тестировании или в финальных запусках, чтобы не засорять логи.  
**Рекомендуется всегда включать при разработке и отладке.**

### Почему trust_remote_code=True  важно?

`trust_remote_code=True` говорит Transformers «разреши подгрузить и выполнить кастомный Python-код из репозитория модели на Hugging Face». Для семейств Qwen это часто нужно, потому что у них есть собственные классы модели/токенайзера и логика чата, которых может не быть (или они отличаются) в вашей локальной версии `transformers`.


1. Что именно подгружается

* Класс модели (например, `QWenLMHeadModel`) с нестандартной реализацией attention/ROPE, kv-cache и т.п.
* Кастомные методы генерации (стоп-токены, обработка системных/ролевых сообщений).
* Шаблон диалога/чат-темплейт и вспомогательные функции токенайзера, чтобы `apply_chat_template` формировал промпт в «родном» формате Qwen.
  Без этого `AutoModelForCausalLM` попытается использовать «ближайший» стандартный класс и:
* либо упадёт с ошибкой «неизвестный тип модели»,
* либо загрузится, но будет генерировать «не по канонам» Qwen (стоп-условия, формат чата, спец-токены и т.д. — всё это может поехать).

2. Почему это особенно касается Qwen/Qwen1.5
   Исторически Qwen выкладывали веса с собственным `modeling_qwen.py` и особыми конфигами. Поддержка в `transformers` догоняет не сразу: если у вас версия библиотеки ниже той, где данная ветка Qwen полностью встроена, без `trust_remote_code` модель не поднимется корректно.

3. Когда можно НЕ ставить `trust_remote_code=True`

* Если у вас свежий `transformers`, где **конкретно ваш чекпойнт** уже поддержан «из коробки». Проверка:

  ```python
  from transformers import AutoConfig
  cfg = AutoConfig.from_pretrained("Qwen/Qwen1.5-7B-Chat")
  print(cfg.model_type)  # должно быть 'qwen' или соответствующий поддерживаемый тип
  ```

  И попробуйте загрузить без `trust_remote_code`. Если загружается и чат работает (правильно форматируются роли/стоп-токены) — можно оставить `False`.

4. Соображения безопасности
   `trust_remote_code=True` исполняет **чужой** код из репозитория модели.

* Зафиксируйте ревизию: `revision="main@<commit_sha>"` или хэш тега.
* Просмотрите `modeling_*.py`, `tokenization_*.py`, `configuration_*.py`.
* После первого скачивания можно запускать с `local_files_only=True`.
* Держите `transformers` и `safetensors` актуальными.

5. Типичные симптомы, если поставить `False`, когда нужен `True`

* Ошибка вида `ValueError: Unrecognized model in Qwen...` или `AutoModelForCausalLM` не знает `qwen`.
* Модель грузится, но `apply_chat_template` даёт странный промпт/падает.
* Генерация обрывается/игнорирует системные роли или «забывает» про специальные стоп-токены.

Итого: для `Qwen/Qwen1.5-7B-Chat` в большинстве окружений действительно **правильно** ставить `trust_remote_code=True`, пока вы не убедились, что ваша версия `transformers` уже полностью поддерживает именно этот релиз Qwen без удалённого кода.


### **Тестирование модели Qwen 1.5 7B**

## **Подготовка среды Colab для тестирования следующей модели**

При подготовке среды Colab для тестирования следующей модели проводим:
### **Перезапуск + Очистка + Инициализация**
---
 Почему мы перезапускаем среду Colab перед тестированием каждой новой LLM-модели?

При сравнении нескольких локальных языковых моделей (LLM) в Google Colab мы сталкиваемся с жёсткими ограничениями:

- **Ограниченный объём оперативной памяти (RAM)**
- **Ограниченный объём видеопамяти (GPU VRAM)**
- **Ограниченное дисковое пространство (обычно ~100 ГБ, но быстро заполняется весами моделей)**

Даже если тестовые данные и настройки для всех моделей одинаковые — **каждая модель требует значительных ресурсов** для загрузки, инициализации и вывода. После тестирования первой модели в памяти остаются:

- Загруженные веса предыдущей модели (иногда гигабайты)
- Кэши, временные файлы, токенизаторы, промежуточные тензоры
- Переменные Python, ссылки на объекты, которые мешают сборщику мусора освободить память

**Всё это приводит к нехватке памяти уже при загрузке второй модели — даже если код написан идеально.**

---

 **Как мы решаем эту проблему?**

Мы **принудительно перезапускаем среду Colab** перед тестированием каждой новой модели.

Это даёт нам:

-  **Полная очистка RAM и VRAM** — все объекты Python, веса моделей, кэши GPU удаляются.
-  **Чистое Python-окружение** — нет остатков переменных, импортов или состояний от предыдущей модели.
-  **Гарантия, что каждая модель тестируется в одинаковых условиях** — без влияния "прошлого".

---

 **Почему мы дополнительно очищаем диск вручную?**

При перезапуске среды (runtime) **файловая система `/content/` НЕ очищается автоматически**. А мы:

- Скачиваем веса моделей (иногда по 5–20 ГБ на каждую!)
- Сохраняем промежуточные результаты, логи, кэши токенизаторов
- Распаковываем датасеты и вспомогательные файлы

Если не очистить диск — **место быстро заканчивается**, и следующая модель просто не сможет загрузиться или распаковаться. Поэтому мы **вручную очищаем `/content/`** перед загрузкой следующей модели — чтобы освободить место и начать "с чистого диска".

**Инициализация среды Colab**

````python
# @title
# ================================================================
# Инициализация среды Colab:
#
# Установка библиотек
# Вспомогательные функции для автоматизации тестов
# Служебные функции для работы с результатами (сохранение / загрузка)
# Универсальная функция тестирования модели
# Авторизация в Hugging Face
# ================================================================

# установка библиотек
!pip install -q transformers torch sentencepiece pandas

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import pandas as pd
import time
from IPython.display import display
from google.colab import drive
import os
import shutil
from google.colab import userdata
from huggingface_hub import login


# ===========================================
#  ГЛОБАЛЬНЫЕ СТРУКТУРЫ И ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ ДЛЯ ТЕСТИРОВАНИЯ МОДЕЛЕЙ
# ===========================================

# --------------------------------------------------
#  1. Фиксированные тестовые вопросы
# --------------------------------------------------

TEST_QUESTIONS = [
    "Термометр показывает плюс 15 градусов. Сколько градусов покажут два таких термометра?",
    "На дереве сидели 7 воробьёв, одного из них съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

# --------------------------------------------------
#  2. Автоматически формируем messages для каждого вопроса
# --------------------------------------------------

TEST_MESSAGES = [
    [{"role": "user", "content": q}] for q in TEST_QUESTIONS
]

# --------------------------------------------------
#  3. Глобальный список вопросов для цикла в функциях тестирования
# Структура: [(номер, messages, текст_вопроса), ...]
# Номера генерируются автоматически
# --------------------------------------------------

TEST_CASES = [
    (i + 1, TEST_MESSAGES[i], TEST_QUESTIONS[i])
    for i in range(len(TEST_QUESTIONS))
]

print(f"✅ Загружено {len(TEST_CASES)} тестовых вопросов.")


# --------------------------------------------------
# 4. Универсальный системный промпт для всех моделей
#  Используется в test_model_universal() для честного и воспроизводимого сравнения
#  Английский язык выбран потому, что:
#    - Все Instruct-модели (Saiga, Mistral, Qwen, DeepSeek, OpenChat) обучены на инструкциях на английском.
#    - Не влияет на язык ответа — модель сама отвечает на языке вопроса.
#    - Гарантирует одинаковые условия тестирования для всех моделей.
# --------------------------------------------------

SYSTEM_PROMPT = "You are a helpful, precise, and concise assistant."

print(f"✅ Системный промпт установлен: \"{SYSTEM_PROMPT}\"")


# --------------------------------------------------
#  5. Динамическая инициализация таблицы результатов
#  Автоматически создаёт колонки под любое количество вопросов
# --------------------------------------------------

def init_results_table():
    """
    Инициализирует или возвращает существующую таблицу результатов.
    Колонки динамически формируются под количество вопросов в TEST_CASES.
    """
    global results_df

    # Если таблица уже существует — не пересоздаём
    if 'results_df' in globals() and isinstance(results_df, pd.DataFrame) and not results_df.empty:
        print("ℹ️ Таблица результатов уже существует — оставляем как есть.")
        return results_df

    # Формируем названия колонок динамически — БЕЗ обрезки текста вопроса
    columns = ['Модель']

    for i, (_, _, question_text) in enumerate(TEST_CASES, 1):
        columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])

    columns.append('Общее время (сек)')
    columns.append('Итоговый вывод')  # ← для ручного заполнения после анализа

    # Создаём пустую таблицу
    results_df = pd.DataFrame(columns=columns)
    print(f"✅ Таблица результатов инициализирована под {len(TEST_CASES)} вопросов.")
    return results_df


# Инициализируем таблицу
results_df = init_results_table()


# --------------------------------------------------
#  6. УНИВЕРСАЛЬНАЯ функция добавления результатов в таблицу
#  Работает с ЛЮБЫМ количеством вопросов
# --------------------------------------------------

def add_to_results(model_name, answers, times, total_time):
    """
    УНИВЕРСАЛЬНАЯ функция: добавляет результаты тестирования модели в таблицу.
    Поддерживает ЛЮБОЕ количество вопросов.
    Итоговый вывод остаётся пустым — его заполняет пользователь вручную после анализа.

     Параметры:
        model_name : str
            Название модели (например, "Qwen/Qwen1.5-7B-Chat").
        answers : list[str]
            Список сгенерированных ответов — по одному на каждый вопрос.
        times : list[float]
            Список времён генерации — по одному на каждый ответ.
        total_time : float
            Общее время выполнения всех запросов.

     Пример вызова:
        result = test_model_universal(tokenizer, model)
        add_to_results("Saiga-7B", result["answers"], result["times"], result["total_time"])
    """
    global results_df

    # Проверка соответствия длины списков
    if len(answers) != len(times):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а времён {len(times)}")
    if len(answers) != len(TEST_QUESTIONS):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а вопросов {len(TEST_QUESTIONS)}")

    # Создаём новую строку результата
    new_row = {'Модель': model_name}

    # Заполняем данные по каждому вопросу
    for i in range(len(answers)):
        q_num = i + 1

        # Ищем точное имя колонки
        question_col = f"Вопрос {q_num}"
        if question_col not in results_df.columns:
            raise KeyError(f"Не найдена колонка для вопроса {q_num}: '{question_col}'")

        # Заполняем текст вопроса, ответ и время
        new_row[question_col] = TEST_QUESTIONS[i]
        new_row[f"Ответ {q_num}"] = answers[i]
        new_row[f"Время ответа {q_num} (сек)"] = round(times[i], 2)

    # Заполняем общее время и оставляем итог пустым
    new_row['Общее время (сек)'] = round(total_time, 2)
    new_row['Итоговый вывод'] = ""

    # Добавляем строку в DataFrame
    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)
    print(f"✅ Результаты для {model_name} добавлены в таблицу. Итоговый вывод — заполните вручную.")

# --------------------------------------------------
# 7. Функция добавления итогового вывода по модели
# --------------------------------------------------

def add_final_conclusion(model_name, conclusion):
    """
    Ручное добавление итогового анализа для указанной модели.
    Обновляет существующую строку в таблице results_df.

    Параметры:
        model_name : str
            Название модели, как оно было добавлено ранее (например, "Qwen/Qwen1.5-7B-Chat").
        conclusion : str
            Текст итогового вывода — анализ логики, ошибок, скорости и т.д.

    """
    global results_df

    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов не инициализирована.")
        return

    # Находим индекс строки с указанной моделью
    mask = results_df['Модель'] == model_name
    if not mask.any():
        print(f"❌ Модель '{model_name}' не найдена в таблице.")
        return

    # Обновляем значение в колонке "Итоговый вывод"
    results_df.loc[mask, 'Итоговый вывод'] = conclusion
    print(f"✅ Итоговый вывод для модели '{model_name}' успешно добавлен.")


# --------------------------------------------------
#  8. Функция вывода результатов (улучшенная для сравнения моделей)
# --------------------------------------------------

def show_results():
    """
    Выводит таблицу сравнения моделей с полными ответами и удобным форматированием.
    """
    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов пуста или не инициализирована.")
        return

    print("\n📊 СРАВНЕНИЕ МОДЕЛЕЙ НА ТЕСТОВЫХ ВОПРОСАХ\n")

    # Настройка pandas для полного отображения текста
    pd.set_option('display.max_colwidth', None)
    pd.set_option('display.width', None)
    pd.set_option('display.max_columns', None)

    # Определяем колонки для отображения
    cols_to_show = ['Модель']
    for i in range(1, len(TEST_QUESTIONS) + 1):
        cols_to_show.extend([f'Вопрос {i}', f'Ответ {i}'])
    cols_to_show.append('Общее время (сек)')
    cols_to_show.append('Итоговый вывод')

    # Проверка наличия колонок
    missing = [col for col in cols_to_show if col not in results_df.columns]
    if missing:
        print(f"⚠️ Отсутствуют колонки: {missing}. Возможно, таблица не инициализирована корректно.")
        display(results_df)
        return

    # Формируем сводную таблицу
    summary_df = results_df[cols_to_show].copy()

    # Стилизация
    styled = summary_df.style.set_properties(**{
        'text-align': 'left',
        'white-space': 'pre-wrap',
        'word-wrap': 'break-word',
        'vertical-align': 'top',
        'font-family': 'Consolas, monospace',
        'font-size': '13px'
    }).set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#f0f0f0'), ('font-weight', 'bold')]},
        {'selector': 'td', 'props': [('border', '1px solid #ddd'), ('padding', '10px')]}
    ]).set_caption("📈 Сравнение локальных LLM на тестах логики и креативности")

    display(styled)



# ============================================
#  УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ LLM
# ============================================

# --- ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ

def _infer_device(model):
    """Определяет, на каком устройстве (CPU/GPU) работает модель."""
    try:
        return next(model.parameters()).device
    except StopIteration:
        return torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")

def _ensure_padding_ids(tokenizer, model):
    """Гарантирует, что pad_token_id задан — иначе используем eos_token_id."""
    if tokenizer.pad_token_id is None and tokenizer.eos_token_id is not None:
        tokenizer.pad_token_id = tokenizer.eos_token_id

def _maybe_inject_system(messages, system_prompt):
    """Добавляет системный промпт в начало, если его нет и он задан."""
    if not system_prompt:
        return list(messages)
    if any(m.get("role") == "system" for m in messages):
        return list(messages)
    return [{"role": "system", "content": system_prompt}] + list(messages)

def _build_fallback_prompt(messages, system_prompt=None):
    """Создаёт простой текстовый промпт, если у токенизатора нет chat_template."""
    parts = []
    if system_prompt:
        parts.append(f"system\n{system_prompt}\n")
    for m in messages:
        parts.append(f"{m.get('role', 'user')}\n{m.get('content', '')}\n")
    parts.append("assistant\n")
    return "".join(parts)


# --- ОСНОВНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ ---

def test_model_universal(
    tokenizer,
    model,
    device=None,
    *,
    system_prompt=None,
    generation_config=None,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    show_prompts=True
):
    """
    Тестирует модель на всех вопросах из TEST_CASES.
    Возвращает структурированный словарь с ответами, временами и общим временем.
    Подходит для любых моделей и любого числа вопросов.
    """

    # Подготавливаем окружение
    _ensure_padding_ids(tokenizer, model)
    if device is None:
        device = _infer_device(model)

    answers, times = [], []
    total_time = 0.0

    # Если нет вопросов — возвращаем пустой результат
    if not TEST_CASES:
        print("⚠️  Нет тестовых вопросов — возвращаем пустой результат.")
        return {"answers": [], "times": [], "total_time": 0.0}

    # Обрабатываем каждый вопрос
    for question_num, messages, question_text in TEST_CASES:
        print(f"🔹 Вопрос {question_num}: {question_text[:80]}...")

        # Формируем сообщения с системным промптом (если нужен)
        msgs = _maybe_inject_system(messages, system_prompt)

        # Генерируем промпт — через шаблон или fallback
        try:
            prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        except (ValueError, AttributeError, NotImplementedError):
            print("⚠️  Без chat_template — используем fallback-формат.")
            prompt = _build_fallback_prompt(msgs, system_prompt=system_prompt)

        if show_prompts:
            print(f"📝 Промпт:\n{prompt}\n{'='*60}")

        # Токенизируем и переносим на устройство
        inputs = tokenizer(prompt, return_tensors="pt", add_special_tokens=False)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        # Генерируем ответ
        start_time = time.time()
        with torch.no_grad():
            if generation_config is not None:
                # Используем переданный конфиг генерации
                gc = generation_config
                if getattr(gc, "pad_token_id", None) is None and tokenizer.pad_token_id is not None:
                    gc = gc.clone()
                    gc.pad_token_id = tokenizer.pad_token_id
                outputs = model.generate(**inputs, generation_config=gc)
            else:
                # Используем стандартные параметры
                outputs = model.generate(
                    **inputs,
                    max_new_tokens=max_new_tokens,
                    do_sample=True,
                    temperature=temperature,
                    top_p=top_p,
                    pad_token_id=tokenizer.pad_token_id,
                    eos_token_id=getattr(model.config, "eos_token_id", None)
                )

        gen_time = time.time() - start_time
        total_time += gen_time

        # Декодируем только сгенерированную часть (после входного промпта)
        out_ids = outputs[0]
        in_len = inputs["input_ids"].shape[1]
        gen_ids = out_ids[in_len:]
        answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

        # Выводим результат
        print(f"✅ Ответ {question_num} за {gen_time:.2f} сек")
        print(f"💬 {answer}\n")

        answers.append(answer)
        times.append(gen_time)

    # Возвращаем всё в структурированном виде — для любого числа вопросов
    return {
        "answers": answers,
        "times": times,
        "total_time": total_time
    }


# --------------------------------------------------
#  1. Путь по умолчанию для сохранения результатов на Google Drive
#  Легко переопределить при вызове функций
# --------------------------------------------------

DEFAULT_SAVE_PATH = "/content/drive/MyDrive/llm_benchmark/llm_comparison_results.csv"


# --------------------------------------------------
#  2. Умная загрузка результатов с адаптацией структуры
#  Загружает CSV и автоматически приводит к актуальной схеме вопросов
#  Если файл отсутствует — создаёт новую таблицу
#  Гарантирует совместимость с add_to_results()
# --------------------------------------------------

def load_or_init_results(test_questions=TEST_QUESTIONS, filepath=DEFAULT_SAVE_PATH):
    """
    Загружает сохранённые результаты из CSV и адаптирует структуру под текущие вопросы.
    Если файла нет — создаёт новую таблицу.
    Возвращает DataFrame `results_df`.

    Параметры:
        test_questions : list — список вопросов (по умолчанию TEST_QUESTIONS)
        filepath : str — путь к файлу CSV

    Использование:
        results_df = load_or_init_results()
    """
    # Формируем ожидаемую структуру колонок на основе текущих вопросов
    expected_columns = ['Модель']
    for i in range(1, len(test_questions) + 1):
        expected_columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])
    expected_columns.extend(['Общее время (сек)', 'Итоговый вывод'])

    #  Проверяем наличие файла
    if os.path.exists(filepath):
        try:
            #  Загружаем данные
            loaded_df = pd.read_csv(filepath)
            print(f"📂 Загружены предыдущие результаты из: {filepath}")

            #  Проверяем, совпадает ли структура
            if set(loaded_df.columns) == set(expected_columns):
                results_df = loaded_df.copy()
                print(f"✅ Структура актуальна. В таблице {len(results_df)} моделей.")
            else:
                # Адаптируем старую таблицу под новую структуру
                print("⚠️ Структура устарела — создаём обновлённую версию...")
                results_df = pd.DataFrame(columns=expected_columns)

                #  Переносим совместимые данные
                for _, row in loaded_df.iterrows():
                    new_row = {}
                    # Общие поля
                    for col in ['Модель', 'Общее время (сек)', 'Итоговый вывод']:
                        new_row[col] = row.get(col, "")
                    # Поля по каждому вопросу
                    for i in range(1, len(test_questions) + 1):
                        for ft in ['Вопрос', 'Ответ']:
                            new_row[f"{ft} {i}"] = row.get(f"{ft} {i}", "")
                        new_row[f"Время ответа {i} (сек)"] = row.get(f"Время ответа {i} (сек)", 0.0)
                    # Добавляем строку
                    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)

                print(f"✅ Создана обновлённая таблица с {len(results_df)} перенесёнными записями.")

        except Exception as e:
            # При ошибке — создаём новую таблицу
            print(f"❌ Ошибка загрузки: {e}. Создаём новую таблицу.")
            results_df = pd.DataFrame(columns=expected_columns)
    else:
        # Файл не найден — инициализируем с нуля
        print("🆕 Файл не найден — создаём новую таблицу.")
        results_df = pd.DataFrame(columns=expected_columns)
        print(f"✅ Таблица инициализирована под {len(test_questions)} вопросов.")

    # Сбрасываем индекс для чистоты
    return results_df.reset_index(drop=True)


# --------------------------------------------------
# 3. Надёжное сохранение чекпоинта на Google Drive
#  Создаёт папку, если её нет
#  Кодировка utf-8-sig для корректного открытия в Excel
#  Выводит подтверждение — можно перезапускать среду
# --------------------------------------------------

def save_results_checkpoint(df, filepath=DEFAULT_SAVE_PATH):
    """
    Сохраняет DataFrame в CSV на Google Drive.

    Параметры:
        df : pd.DataFrame — таблица для сохранения
        filepath : str — путь к файлу CSV

     Использование:
        save_results_checkpoint(results_df)
    """
    #  Создаём директорию, если не существует
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    #  Сохраняем с корректной кодировкой
    df.to_csv(filepath, index=False, encoding='utf-8-sig')
    print(f"💾 Чекпоинт сохранён: {filepath}")
    print("✅ Теперь можно перезапустить среду.")


# Авторизация в Hugging Face
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token)
print("✅ Авторизованы в Hugging Face")
````

````python
# @title
# ================================================================
# Инициализация среды Colab:
#
# Установка библиотек
# Вспомогательные функции для автоматизации тестов
# Служебные функции для работы с результатами (сохранение / загрузка)
# Универсальная функция тестирования модели
# Авторизация в Hugging Face
# ================================================================

# установка библиотек
!pip install -q transformers torch sentencepiece pandas

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import pandas as pd
import time
from IPython.display import display
from google.colab import drive
import os
import shutil
from google.colab import userdata
from huggingface_hub import login


# ===========================================
#  ГЛОБАЛЬНЫЕ СТРУКТУРЫ И ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ ДЛЯ ТЕСТИРОВАНИЯ МОДЕЛЕЙ
# ===========================================

# --------------------------------------------------
#  1. Фиксированные тестовые вопросы
# --------------------------------------------------

TEST_QUESTIONS = [
    "Термометр показывает плюс 15 градусов. Сколько градусов покажут два таких термометра?",
    "На дереве сидели 7 воробьёв, одного из них съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

# --------------------------------------------------
#  2. Автоматически формируем messages для каждого вопроса
# --------------------------------------------------

TEST_MESSAGES = [
    [{"role": "user", "content": q}] for q in TEST_QUESTIONS
]

# --------------------------------------------------
#  3. Глобальный список вопросов для цикла в функциях тестирования
# Структура: [(номер, messages, текст_вопроса), ...]
# Номера генерируются автоматически
# --------------------------------------------------

TEST_CASES = [
    (i + 1, TEST_MESSAGES[i], TEST_QUESTIONS[i])
    for i in range(len(TEST_QUESTIONS))
]

print(f"✅ Загружено {len(TEST_CASES)} тестовых вопросов.")


# --------------------------------------------------
# 4. Универсальный системный промпт для всех моделей
#  Используется в test_model_universal() для честного и воспроизводимого сравнения
#  Английский язык выбран потому, что:
#    - Все Instruct-модели (Saiga, Mistral, Qwen, DeepSeek, OpenChat) обучены на инструкциях на английском.
#    - Не влияет на язык ответа — модель сама отвечает на языке вопроса.
#    - Гарантирует одинаковые условия тестирования для всех моделей.
# --------------------------------------------------

SYSTEM_PROMPT = "You are a helpful, precise, and concise assistant."

print(f"✅ Системный промпт установлен: \"{SYSTEM_PROMPT}\"")


# --------------------------------------------------
#  5. Динамическая инициализация таблицы результатов
#  Автоматически создаёт колонки под любое количество вопросов
# --------------------------------------------------

def init_results_table():
    """
    Инициализирует или возвращает существующую таблицу результатов.
    Колонки динамически формируются под количество вопросов в TEST_CASES.
    """
    global results_df

    # Если таблица уже существует — не пересоздаём
    if 'results_df' in globals() and isinstance(results_df, pd.DataFrame) and not results_df.empty:
        print("ℹ️ Таблица результатов уже существует — оставляем как есть.")
        return results_df

    # Формируем названия колонок динамически — БЕЗ обрезки текста вопроса
    columns = ['Модель']

    for i, (_, _, question_text) in enumerate(TEST_CASES, 1):
        columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])

    columns.append('Общее время (сек)')
    columns.append('Итоговый вывод')  # ← для ручного заполнения после анализа

    # Создаём пустую таблицу
    results_df = pd.DataFrame(columns=columns)
    print(f"✅ Таблица результатов инициализирована под {len(TEST_CASES)} вопросов.")
    return results_df


# Инициализируем таблицу
results_df = init_results_table()


# --------------------------------------------------
#  6. УНИВЕРСАЛЬНАЯ функция добавления результатов в таблицу
#  Работает с ЛЮБЫМ количеством вопросов
# --------------------------------------------------

def add_to_results(model_name, answers, times, total_time):
    """
    УНИВЕРСАЛЬНАЯ функция: добавляет результаты тестирования модели в таблицу.
    Поддерживает ЛЮБОЕ количество вопросов.
    Итоговый вывод остаётся пустым — его заполняет пользователь вручную после анализа.

     Параметры:
        model_name : str
            Название модели (например, "Qwen/Qwen1.5-7B-Chat").
        answers : list[str]
            Список сгенерированных ответов — по одному на каждый вопрос.
        times : list[float]
            Список времён генерации — по одному на каждый ответ.
        total_time : float
            Общее время выполнения всех запросов.

     Пример вызова:
        result = test_model_universal(tokenizer, model)
        add_to_results("Saiga-7B", result["answers"], result["times"], result["total_time"])
    """
    global results_df

    # Проверка соответствия длины списков
    if len(answers) != len(times):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а времён {len(times)}")
    if len(answers) != len(TEST_QUESTIONS):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а вопросов {len(TEST_QUESTIONS)}")

    # Создаём новую строку результата
    new_row = {'Модель': model_name}

    # Заполняем данные по каждому вопросу
    for i in range(len(answers)):
        q_num = i + 1

        # Ищем точное имя колонки
        question_col = f"Вопрос {q_num}"
        if question_col not in results_df.columns:
            raise KeyError(f"Не найдена колонка для вопроса {q_num}: '{question_col}'")

        # Заполняем текст вопроса, ответ и время
        new_row[question_col] = TEST_QUESTIONS[i]
        new_row[f"Ответ {q_num}"] = answers[i]
        new_row[f"Время ответа {q_num} (сек)"] = round(times[i], 2)

    # Заполняем общее время и оставляем итог пустым
    new_row['Общее время (сек)'] = round(total_time, 2)
    new_row['Итоговый вывод'] = ""

    # Добавляем строку в DataFrame
    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)
    print(f"✅ Результаты для {model_name} добавлены в таблицу. Итоговый вывод — заполните вручную.")

# --------------------------------------------------
# 7. Функция добавления итогового вывода по модели
# --------------------------------------------------

def add_final_conclusion(model_name, conclusion):
    """
    Ручное добавление итогового анализа для указанной модели.
    Обновляет существующую строку в таблице results_df.

    Параметры:
        model_name : str
            Название модели, как оно было добавлено ранее (например, "Qwen/Qwen1.5-7B-Chat").
        conclusion : str
            Текст итогового вывода — анализ логики, ошибок, скорости и т.д.

    """
    global results_df

    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов не инициализирована.")
        return

    # Находим индекс строки с указанной моделью
    mask = results_df['Модель'] == model_name
    if not mask.any():
        print(f"❌ Модель '{model_name}' не найдена в таблице.")
        return

    # Обновляем значение в колонке "Итоговый вывод"
    results_df.loc[mask, 'Итоговый вывод'] = conclusion
    print(f"✅ Итоговый вывод для модели '{model_name}' успешно добавлен.")


# --------------------------------------------------
#  8. Функция вывода результатов (улучшенная для сравнения моделей)
# --------------------------------------------------

def show_results():
    """
    Выводит таблицу сравнения моделей с полными ответами и удобным форматированием.
    """
    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов пуста или не инициализирована.")
        return

    print("\n📊 СРАВНЕНИЕ МОДЕЛЕЙ НА ТЕСТОВЫХ ВОПРОСАХ\n")

    # Настройка pandas для полного отображения текста
    pd.set_option('display.max_colwidth', None)
    pd.set_option('display.width', None)
    pd.set_option('display.max_columns', None)

    # Определяем колонки для отображения
    cols_to_show = ['Модель']
    for i in range(1, len(TEST_QUESTIONS) + 1):
        cols_to_show.extend([f'Вопрос {i}', f'Ответ {i}'])
    cols_to_show.append('Общее время (сек)')
    cols_to_show.append('Итоговый вывод')

    # Проверка наличия колонок
    missing = [col for col in cols_to_show if col not in results_df.columns]
    if missing:
        print(f"⚠️ Отсутствуют колонки: {missing}. Возможно, таблица не инициализирована корректно.")
        display(results_df)
        return

    # Формируем сводную таблицу
    summary_df = results_df[cols_to_show].copy()

    # Стилизация
    styled = summary_df.style.set_properties(**{
        'text-align': 'left',
        'white-space': 'pre-wrap',
        'word-wrap': 'break-word',
        'vertical-align': 'top',
        'font-family': 'Consolas, monospace',
        'font-size': '13px'
    }).set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#f0f0f0'), ('font-weight', 'bold')]},
        {'selector': 'td', 'props': [('border', '1px solid #ddd'), ('padding', '10px')]}
    ]).set_caption("📈 Сравнение локальных LLM на тестах логики и креативности")

    display(styled)



# ============================================
#  УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ LLM
# ============================================

# --- ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ

def _infer_device(model):
    """Определяет, на каком устройстве (CPU/GPU) работает модель."""
    try:
        return next(model.parameters()).device
    except StopIteration:
        return torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")

def _ensure_padding_ids(tokenizer, model):
    """Гарантирует, что pad_token_id задан — иначе используем eos_token_id."""
    if tokenizer.pad_token_id is None and tokenizer.eos_token_id is not None:
        tokenizer.pad_token_id = tokenizer.eos_token_id

def _maybe_inject_system(messages, system_prompt):
    """Добавляет системный промпт в начало, если его нет и он задан."""
    if not system_prompt:
        return list(messages)
    if any(m.get("role") == "system" for m in messages):
        return list(messages)
    return [{"role": "system", "content": system_prompt}] + list(messages)

def _build_fallback_prompt(messages, system_prompt=None):
    """Создаёт простой текстовый промпт, если у токенизатора нет chat_template."""
    parts = []
    if system_prompt:
        parts.append(f"system\n{system_prompt}\n")
    for m in messages:
        parts.append(f"{m.get('role', 'user')}\n{m.get('content', '')}\n")
    parts.append("assistant\n")
    return "".join(parts)


# --- ОСНОВНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ ---

def test_model_universal(
    tokenizer,
    model,
    device=None,
    *,
    system_prompt=None,
    generation_config=None,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    show_prompts=True
):
    """
    Тестирует модель на всех вопросах из TEST_CASES.
    Возвращает структурированный словарь с ответами, временами и общим временем.
    Подходит для любых моделей и любого числа вопросов.
    """

    # Подготавливаем окружение
    _ensure_padding_ids(tokenizer, model)
    if device is None:
        device = _infer_device(model)

    answers, times = [], []
    total_time = 0.0

    # Если нет вопросов — возвращаем пустой результат
    if not TEST_CASES:
        print("⚠️  Нет тестовых вопросов — возвращаем пустой результат.")
        return {"answers": [], "times": [], "total_time": 0.0}

    # Обрабатываем каждый вопрос
    for question_num, messages, question_text in TEST_CASES:
        print(f"🔹 Вопрос {question_num}: {question_text[:80]}...")

        # Формируем сообщения с системным промптом (если нужен)
        msgs = _maybe_inject_system(messages, system_prompt)

        # Генерируем промпт — через шаблон или fallback
        try:
            prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        except (ValueError, AttributeError, NotImplementedError):
            print("⚠️  Без chat_template — используем fallback-формат.")
            prompt = _build_fallback_prompt(msgs, system_prompt=system_prompt)

        if show_prompts:
            print(f"📝 Промпт:\n{prompt}\n{'='*60}")

        # Токенизируем и переносим на устройство
        inputs = tokenizer(prompt, return_tensors="pt", add_special_tokens=False)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        # Генерируем ответ
        start_time = time.time()
        with torch.no_grad():
            if generation_config is not None:
                # Используем переданный конфиг генерации
                gc = generation_config
                if getattr(gc, "pad_token_id", None) is None and tokenizer.pad_token_id is not None:
                    gc = gc.clone()
                    gc.pad_token_id = tokenizer.pad_token_id
                outputs = model.generate(**inputs, generation_config=gc)
            else:
                # Используем стандартные параметры
                outputs = model.generate(
                    **inputs,
                    max_new_tokens=max_new_tokens,
                    do_sample=True,
                    temperature=temperature,
                    top_p=top_p,
                    pad_token_id=tokenizer.pad_token_id,
                    eos_token_id=getattr(model.config, "eos_token_id", None)
                )

        gen_time = time.time() - start_time
        total_time += gen_time

        # Декодируем только сгенерированную часть (после входного промпта)
        out_ids = outputs[0]
        in_len = inputs["input_ids"].shape[1]
        gen_ids = out_ids[in_len:]
        answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

        # Выводим результат
        print(f"✅ Ответ {question_num} за {gen_time:.2f} сек")
        print(f"💬 {answer}\n")

        answers.append(answer)
        times.append(gen_time)

    # Возвращаем всё в структурированном виде — для любого числа вопросов
    return {
        "answers": answers,
        "times": times,
        "total_time": total_time
    }


# --------------------------------------------------
#  1. Путь по умолчанию для сохранения результатов на Google Drive
#  Легко переопределить при вызове функций
# --------------------------------------------------

DEFAULT_SAVE_PATH = "/content/drive/MyDrive/llm_benchmark/llm_comparison_results.csv"


# --------------------------------------------------
#  2. Умная загрузка результатов с адаптацией структуры
#  Загружает CSV и автоматически приводит к актуальной схеме вопросов
#  Если файл отсутствует — создаёт новую таблицу
#  Гарантирует совместимость с add_to_results()
# --------------------------------------------------

def load_or_init_results(test_questions=TEST_QUESTIONS, filepath=DEFAULT_SAVE_PATH):
    """
    Загружает сохранённые результаты из CSV и адаптирует структуру под текущие вопросы.
    Если файла нет — создаёт новую таблицу.
    Возвращает DataFrame `results_df`.

    Параметры:
        test_questions : list — список вопросов (по умолчанию TEST_QUESTIONS)
        filepath : str — путь к файлу CSV

    Использование:
        results_df = load_or_init_results()
    """
    # Формируем ожидаемую структуру колонок на основе текущих вопросов
    expected_columns = ['Модель']
    for i in range(1, len(test_questions) + 1):
        expected_columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])
    expected_columns.extend(['Общее время (сек)', 'Итоговый вывод'])

    #  Проверяем наличие файла
    if os.path.exists(filepath):
        try:
            #  Загружаем данные
            loaded_df = pd.read_csv(filepath)
            print(f"📂 Загружены предыдущие результаты из: {filepath}")

            #  Проверяем, совпадает ли структура
            if set(loaded_df.columns) == set(expected_columns):
                results_df = loaded_df.copy()
                print(f"✅ Структура актуальна. В таблице {len(results_df)} моделей.")
            else:
                # Адаптируем старую таблицу под новую структуру
                print("⚠️ Структура устарела — создаём обновлённую версию...")
                results_df = pd.DataFrame(columns=expected_columns)

                #  Переносим совместимые данные
                for _, row in loaded_df.iterrows():
                    new_row = {}
                    # Общие поля
                    for col in ['Модель', 'Общее время (сек)', 'Итоговый вывод']:
                        new_row[col] = row.get(col, "")
                    # Поля по каждому вопросу
                    for i in range(1, len(test_questions) + 1):
                        for ft in ['Вопрос', 'Ответ']:
                            new_row[f"{ft} {i}"] = row.get(f"{ft} {i}", "")
                        new_row[f"Время ответа {i} (сек)"] = row.get(f"Время ответа {i} (сек)", 0.0)
                    # Добавляем строку
                    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)

                print(f"✅ Создана обновлённая таблица с {len(results_df)} перенесёнными записями.")

        except Exception as e:
            # При ошибке — создаём новую таблицу
            print(f"❌ Ошибка загрузки: {e}. Создаём новую таблицу.")
            results_df = pd.DataFrame(columns=expected_columns)
    else:
        # Файл не найден — инициализируем с нуля
        print("🆕 Файл не найден — создаём новую таблицу.")
        results_df = pd.DataFrame(columns=expected_columns)
        print(f"✅ Таблица инициализирована под {len(test_questions)} вопросов.")

    # Сбрасываем индекс для чистоты
    return results_df.reset_index(drop=True)


# --------------------------------------------------
# 3. Надёжное сохранение чекпоинта на Google Drive
#  Создаёт папку, если её нет
#  Кодировка utf-8-sig для корректного открытия в Excel
#  Выводит подтверждение — можно перезапускать среду
# --------------------------------------------------

def save_results_checkpoint(df, filepath=DEFAULT_SAVE_PATH):
    """
    Сохраняет DataFrame в CSV на Google Drive.

    Параметры:
        df : pd.DataFrame — таблица для сохранения
        filepath : str — путь к файлу CSV

     Использование:
        save_results_checkpoint(results_df)
    """
    #  Создаём директорию, если не существует
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    #  Сохраняем с корректной кодировкой
    df.to_csv(filepath, index=False, encoding='utf-8-sig')
    print(f"💾 Чекпоинт сохранён: {filepath}")
    print("✅ Теперь можно перезапустить среду.")


# Авторизация в Hugging Face
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token)
print("✅ Авторизованы в Hugging Face")
````

````python
# @title
# ================================================================
# Инициализация среды Colab:
#
# Установка библиотек
# Вспомогательные функции для автоматизации тестов
# Служебные функции для работы с результатами (сохранение / загрузка)
# Универсальная функция тестирования модели
# Авторизация в Hugging Face
# ================================================================

# установка библиотек
!pip install -q transformers torch sentencepiece pandas

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import pandas as pd
import time
from IPython.display import display
from google.colab import drive
import os
import shutil
from google.colab import userdata
from huggingface_hub import login


# ===========================================
#  ГЛОБАЛЬНЫЕ СТРУКТУРЫ И ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ ДЛЯ ТЕСТИРОВАНИЯ МОДЕЛЕЙ
# ===========================================

# --------------------------------------------------
#  1. Фиксированные тестовые вопросы
# --------------------------------------------------

TEST_QUESTIONS = [
    "Термометр показывает плюс 15 градусов. Сколько градусов покажут два таких термометра?",
    "На дереве сидели 7 воробьёв, одного из них съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

# --------------------------------------------------
#  2. Автоматически формируем messages для каждого вопроса
# --------------------------------------------------

TEST_MESSAGES = [
    [{"role": "user", "content": q}] for q in TEST_QUESTIONS
]

# --------------------------------------------------
#  3. Глобальный список вопросов для цикла в функциях тестирования
# Структура: [(номер, messages, текст_вопроса), ...]
# Номера генерируются автоматически
# --------------------------------------------------

TEST_CASES = [
    (i + 1, TEST_MESSAGES[i], TEST_QUESTIONS[i])
    for i in range(len(TEST_QUESTIONS))
]

print(f"✅ Загружено {len(TEST_CASES)} тестовых вопросов.")


# --------------------------------------------------
# 4. Универсальный системный промпт для всех моделей
#  Используется в test_model_universal() для честного и воспроизводимого сравнения
#  Английский язык выбран потому, что:
#    - Все Instruct-модели (Saiga, Mistral, Qwen, DeepSeek, OpenChat) обучены на инструкциях на английском.
#    - Не влияет на язык ответа — модель сама отвечает на языке вопроса.
#    - Гарантирует одинаковые условия тестирования для всех моделей.
# --------------------------------------------------

SYSTEM_PROMPT = "You are a helpful, precise, and concise assistant."

print(f"✅ Системный промпт установлен: \"{SYSTEM_PROMPT}\"")


# --------------------------------------------------
#  5. Динамическая инициализация таблицы результатов
#  Автоматически создаёт колонки под любое количество вопросов
# --------------------------------------------------

def init_results_table():
    """
    Инициализирует или возвращает существующую таблицу результатов.
    Колонки динамически формируются под количество вопросов в TEST_CASES.
    """
    global results_df

    # Если таблица уже существует — не пересоздаём
    if 'results_df' in globals() and isinstance(results_df, pd.DataFrame) and not results_df.empty:
        print("ℹ️ Таблица результатов уже существует — оставляем как есть.")
        return results_df

    # Формируем названия колонок динамически — БЕЗ обрезки текста вопроса
    columns = ['Модель']

    for i, (_, _, question_text) in enumerate(TEST_CASES, 1):
        columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])

    columns.append('Общее время (сек)')
    columns.append('Итоговый вывод')  # ← для ручного заполнения после анализа

    # Создаём пустую таблицу
    results_df = pd.DataFrame(columns=columns)
    print(f"✅ Таблица результатов инициализирована под {len(TEST_CASES)} вопросов.")
    return results_df


# Инициализируем таблицу
results_df = init_results_table()


# --------------------------------------------------
#  6. УНИВЕРСАЛЬНАЯ функция добавления результатов в таблицу
#  Работает с ЛЮБЫМ количеством вопросов
# --------------------------------------------------

def add_to_results(model_name, answers, times, total_time):
    """
    УНИВЕРСАЛЬНАЯ функция: добавляет результаты тестирования модели в таблицу.
    Поддерживает ЛЮБОЕ количество вопросов.
    Итоговый вывод остаётся пустым — его заполняет пользователь вручную после анализа.

     Параметры:
        model_name : str
            Название модели (например, "Qwen/Qwen1.5-7B-Chat").
        answers : list[str]
            Список сгенерированных ответов — по одному на каждый вопрос.
        times : list[float]
            Список времён генерации — по одному на каждый ответ.
        total_time : float
            Общее время выполнения всех запросов.

     Пример вызова:
        result = test_model_universal(tokenizer, model)
        add_to_results("Saiga-7B", result["answers"], result["times"], result["total_time"])
    """
    global results_df

    # Проверка соответствия длины списков
    if len(answers) != len(times):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а времён {len(times)}")
    if len(answers) != len(TEST_QUESTIONS):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а вопросов {len(TEST_QUESTIONS)}")

    # Создаём новую строку результата
    new_row = {'Модель': model_name}

    # Заполняем данные по каждому вопросу
    for i in range(len(answers)):
        q_num = i + 1

        # Ищем точное имя колонки
        question_col = f"Вопрос {q_num}"
        if question_col not in results_df.columns:
            raise KeyError(f"Не найдена колонка для вопроса {q_num}: '{question_col}'")

        # Заполняем текст вопроса, ответ и время
        new_row[question_col] = TEST_QUESTIONS[i]
        new_row[f"Ответ {q_num}"] = answers[i]
        new_row[f"Время ответа {q_num} (сек)"] = round(times[i], 2)

    # Заполняем общее время и оставляем итог пустым
    new_row['Общее время (сек)'] = round(total_time, 2)
    new_row['Итоговый вывод'] = ""

    # Добавляем строку в DataFrame
    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)
    print(f"✅ Результаты для {model_name} добавлены в таблицу. Итоговый вывод — заполните вручную.")

# --------------------------------------------------
# 7. Функция добавления итогового вывода по модели
# --------------------------------------------------

def add_final_conclusion(model_name, conclusion):
    """
    Ручное добавление итогового анализа для указанной модели.
    Обновляет существующую строку в таблице results_df.

    Параметры:
        model_name : str
            Название модели, как оно было добавлено ранее (например, "Qwen/Qwen1.5-7B-Chat").
        conclusion : str
            Текст итогового вывода — анализ логики, ошибок, скорости и т.д.

    """
    global results_df

    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов не инициализирована.")
        return

    # Находим индекс строки с указанной моделью
    mask = results_df['Модель'] == model_name
    if not mask.any():
        print(f"❌ Модель '{model_name}' не найдена в таблице.")
        return

    # Обновляем значение в колонке "Итоговый вывод"
    results_df.loc[mask, 'Итоговый вывод'] = conclusion
    print(f"✅ Итоговый вывод для модели '{model_name}' успешно добавлен.")


# --------------------------------------------------
#  8. Функция вывода результатов (улучшенная для сравнения моделей)
# --------------------------------------------------

def show_results():
    """
    Выводит таблицу сравнения моделей с полными ответами и удобным форматированием.
    """
    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов пуста или не инициализирована.")
        return

    print("\n📊 СРАВНЕНИЕ МОДЕЛЕЙ НА ТЕСТОВЫХ ВОПРОСАХ\n")

    # Настройка pandas для полного отображения текста
    pd.set_option('display.max_colwidth', None)
    pd.set_option('display.width', None)
    pd.set_option('display.max_columns', None)

    # Определяем колонки для отображения
    cols_to_show = ['Модель']
    for i in range(1, len(TEST_QUESTIONS) + 1):
        cols_to_show.extend([f'Вопрос {i}', f'Ответ {i}'])
    cols_to_show.append('Общее время (сек)')
    cols_to_show.append('Итоговый вывод')

    # Проверка наличия колонок
    missing = [col for col in cols_to_show if col not in results_df.columns]
    if missing:
        print(f"⚠️ Отсутствуют колонки: {missing}. Возможно, таблица не инициализирована корректно.")
        display(results_df)
        return

    # Формируем сводную таблицу
    summary_df = results_df[cols_to_show].copy()

    # Стилизация
    styled = summary_df.style.set_properties(**{
        'text-align': 'left',
        'white-space': 'pre-wrap',
        'word-wrap': 'break-word',
        'vertical-align': 'top',
        'font-family': 'Consolas, monospace',
        'font-size': '13px'
    }).set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#f0f0f0'), ('font-weight', 'bold')]},
        {'selector': 'td', 'props': [('border', '1px solid #ddd'), ('padding', '10px')]}
    ]).set_caption("📈 Сравнение локальных LLM на тестах логики и креативности")

    display(styled)



# ============================================
#  УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ LLM
# ============================================

# --- ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ

def _infer_device(model):
    """Определяет, на каком устройстве (CPU/GPU) работает модель."""
    try:
        return next(model.parameters()).device
    except StopIteration:
        return torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")

def _ensure_padding_ids(tokenizer, model):
    """Гарантирует, что pad_token_id задан — иначе используем eos_token_id."""
    if tokenizer.pad_token_id is None and tokenizer.eos_token_id is not None:
        tokenizer.pad_token_id = tokenizer.eos_token_id

def _maybe_inject_system(messages, system_prompt):
    """Добавляет системный промпт в начало, если его нет и он задан."""
    if not system_prompt:
        return list(messages)
    if any(m.get("role") == "system" for m in messages):
        return list(messages)
    return [{"role": "system", "content": system_prompt}] + list(messages)

def _build_fallback_prompt(messages, system_prompt=None):
    """Создаёт простой текстовый промпт, если у токенизатора нет chat_template."""
    parts = []
    if system_prompt:
        parts.append(f"system\n{system_prompt}\n")
    for m in messages:
        parts.append(f"{m.get('role', 'user')}\n{m.get('content', '')}\n")
    parts.append("assistant\n")
    return "".join(parts)


# --- ОСНОВНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ ---

def test_model_universal(
    tokenizer,
    model,
    device=None,
    *,
    system_prompt=None,
    generation_config=None,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    show_prompts=True
):
    """
    Тестирует модель на всех вопросах из TEST_CASES.
    Возвращает структурированный словарь с ответами, временами и общим временем.
    Подходит для любых моделей и любого числа вопросов.
    """

    # Подготавливаем окружение
    _ensure_padding_ids(tokenizer, model)
    if device is None:
        device = _infer_device(model)

    answers, times = [], []
    total_time = 0.0

    # Если нет вопросов — возвращаем пустой результат
    if not TEST_CASES:
        print("⚠️  Нет тестовых вопросов — возвращаем пустой результат.")
        return {"answers": [], "times": [], "total_time": 0.0}

    # Обрабатываем каждый вопрос
    for question_num, messages, question_text in TEST_CASES:
        print(f"🔹 Вопрос {question_num}: {question_text[:80]}...")

        # Формируем сообщения с системным промптом (если нужен)
        msgs = _maybe_inject_system(messages, system_prompt)

        # Генерируем промпт — через шаблон или fallback
        try:
            prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        except (ValueError, AttributeError, NotImplementedError):
            print("⚠️  Без chat_template — используем fallback-формат.")
            prompt = _build_fallback_prompt(msgs, system_prompt=system_prompt)

        if show_prompts:
            print(f"📝 Промпт:\n{prompt}\n{'='*60}")

        # Токенизируем и переносим на устройство
        inputs = tokenizer(prompt, return_tensors="pt", add_special_tokens=False)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        # Генерируем ответ
        start_time = time.time()
        with torch.no_grad():
            if generation_config is not None:
                # Используем переданный конфиг генерации
                gc = generation_config
                if getattr(gc, "pad_token_id", None) is None and tokenizer.pad_token_id is not None:
                    gc = gc.clone()
                    gc.pad_token_id = tokenizer.pad_token_id
                outputs = model.generate(**inputs, generation_config=gc)
            else:
                # Используем стандартные параметры
                outputs = model.generate(
                    **inputs,
                    max_new_tokens=max_new_tokens,
                    do_sample=True,
                    temperature=temperature,
                    top_p=top_p,
                    pad_token_id=tokenizer.pad_token_id,
                    eos_token_id=getattr(model.config, "eos_token_id", None)
                )

        gen_time = time.time() - start_time
        total_time += gen_time

        # Декодируем только сгенерированную часть (после входного промпта)
        out_ids = outputs[0]
        in_len = inputs["input_ids"].shape[1]
        gen_ids = out_ids[in_len:]
        answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

        # Выводим результат
        print(f"✅ Ответ {question_num} за {gen_time:.2f} сек")
        print(f"💬 {answer}\n")

        answers.append(answer)
        times.append(gen_time)

    # Возвращаем всё в структурированном виде — для любого числа вопросов
    return {
        "answers": answers,
        "times": times,
        "total_time": total_time
    }


# --------------------------------------------------
#  1. Путь по умолчанию для сохранения результатов на Google Drive
#  Легко переопределить при вызове функций
# --------------------------------------------------

DEFAULT_SAVE_PATH = "/content/drive/MyDrive/llm_benchmark/llm_comparison_results.csv"


# --------------------------------------------------
#  2. Умная загрузка результатов с адаптацией структуры
#  Загружает CSV и автоматически приводит к актуальной схеме вопросов
#  Если файл отсутствует — создаёт новую таблицу
#  Гарантирует совместимость с add_to_results()
# --------------------------------------------------

def load_or_init_results(test_questions=TEST_QUESTIONS, filepath=DEFAULT_SAVE_PATH):
    """
    Загружает сохранённые результаты из CSV и адаптирует структуру под текущие вопросы.
    Если файла нет — создаёт новую таблицу.
    Возвращает DataFrame `results_df`.

    Параметры:
        test_questions : list — список вопросов (по умолчанию TEST_QUESTIONS)
        filepath : str — путь к файлу CSV

    Использование:
        results_df = load_or_init_results()
    """
    # Формируем ожидаемую структуру колонок на основе текущих вопросов
    expected_columns = ['Модель']
    for i in range(1, len(test_questions) + 1):
        expected_columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])
    expected_columns.extend(['Общее время (сек)', 'Итоговый вывод'])

    #  Проверяем наличие файла
    if os.path.exists(filepath):
        try:
            #  Загружаем данные
            loaded_df = pd.read_csv(filepath)
            print(f"📂 Загружены предыдущие результаты из: {filepath}")

            #  Проверяем, совпадает ли структура
            if set(loaded_df.columns) == set(expected_columns):
                results_df = loaded_df.copy()
                print(f"✅ Структура актуальна. В таблице {len(results_df)} моделей.")
            else:
                # Адаптируем старую таблицу под новую структуру
                print("⚠️ Структура устарела — создаём обновлённую версию...")
                results_df = pd.DataFrame(columns=expected_columns)

                #  Переносим совместимые данные
                for _, row in loaded_df.iterrows():
                    new_row = {}
                    # Общие поля
                    for col in ['Модель', 'Общее время (сек)', 'Итоговый вывод']:
                        new_row[col] = row.get(col, "")
                    # Поля по каждому вопросу
                    for i in range(1, len(test_questions) + 1):
                        for ft in ['Вопрос', 'Ответ']:
                            new_row[f"{ft} {i}"] = row.get(f"{ft} {i}", "")
                        new_row[f"Время ответа {i} (сек)"] = row.get(f"Время ответа {i} (сек)", 0.0)
                    # Добавляем строку
                    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)

                print(f"✅ Создана обновлённая таблица с {len(results_df)} перенесёнными записями.")

        except Exception as e:
            # При ошибке — создаём новую таблицу
            print(f"❌ Ошибка загрузки: {e}. Создаём новую таблицу.")
            results_df = pd.DataFrame(columns=expected_columns)
    else:
        # Файл не найден — инициализируем с нуля
        print("🆕 Файл не найден — создаём новую таблицу.")
        results_df = pd.DataFrame(columns=expected_columns)
        print(f"✅ Таблица инициализирована под {len(test_questions)} вопросов.")

    # Сбрасываем индекс для чистоты
    return results_df.reset_index(drop=True)


# --------------------------------------------------
# 3. Надёжное сохранение чекпоинта на Google Drive
#  Создаёт папку, если её нет
#  Кодировка utf-8-sig для корректного открытия в Excel
#  Выводит подтверждение — можно перезапускать среду
# --------------------------------------------------

def save_results_checkpoint(df, filepath=DEFAULT_SAVE_PATH):
    """
    Сохраняет DataFrame в CSV на Google Drive.

    Параметры:
        df : pd.DataFrame — таблица для сохранения
        filepath : str — путь к файлу CSV

     Использование:
        save_results_checkpoint(results_df)
    """
    #  Создаём директорию, если не существует
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    #  Сохраняем с корректной кодировкой
    df.to_csv(filepath, index=False, encoding='utf-8-sig')
    print(f"💾 Чекпоинт сохранён: {filepath}")
    print("✅ Теперь можно перезапустить среду.")


# Авторизация в Hugging Face
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token)
print("✅ Авторизованы в Hugging Face")
````

````python
# @title
# ================================================================
# Инициализация среды Colab:
#
# Установка библиотек
# Вспомогательные функции для автоматизации тестов
# Служебные функции для работы с результатами (сохранение / загрузка)
# Универсальная функция тестирования модели
# Авторизация в Hugging Face
# ================================================================

# установка библиотек
!pip install -q transformers torch sentencepiece pandas

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
import pandas as pd
import time
from IPython.display import display
from google.colab import drive
import os
import shutil
from google.colab import userdata
from huggingface_hub import login


# ===========================================
#  ГЛОБАЛЬНЫЕ СТРУКТУРЫ И ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ ДЛЯ ТЕСТИРОВАНИЯ МОДЕЛЕЙ
# ===========================================

# --------------------------------------------------
#  1. Фиксированные тестовые вопросы
# --------------------------------------------------

TEST_QUESTIONS = [
    "Термометр показывает плюс 15 градусов. Сколько градусов покажут два таких термометра?",
    "На дереве сидели 7 воробьёв, одного из них съел кот. Сколько осталось?",
    "По небу летели голубь, муха и сокол. Сколько птиц летело?",
]

# --------------------------------------------------
#  2. Автоматически формируем messages для каждого вопроса
# --------------------------------------------------

TEST_MESSAGES = [
    [{"role": "user", "content": q}] for q in TEST_QUESTIONS
]

# --------------------------------------------------
#  3. Глобальный список вопросов для цикла в функциях тестирования
# Структура: [(номер, messages, текст_вопроса), ...]
# Номера генерируются автоматически
# --------------------------------------------------

TEST_CASES = [
    (i + 1, TEST_MESSAGES[i], TEST_QUESTIONS[i])
    for i in range(len(TEST_QUESTIONS))
]

print(f"✅ Загружено {len(TEST_CASES)} тестовых вопросов.")


# --------------------------------------------------
# 4. Универсальный системный промпт для всех моделей
#  Используется в test_model_universal() для честного и воспроизводимого сравнения
#  Английский язык выбран потому, что:
#    - Все Instruct-модели (Saiga, Mistral, Qwen, DeepSeek, OpenChat) обучены на инструкциях на английском.
#    - Не влияет на язык ответа — модель сама отвечает на языке вопроса.
#    - Гарантирует одинаковые условия тестирования для всех моделей.
# --------------------------------------------------

SYSTEM_PROMPT = "You are a helpful, precise, and concise assistant."

print(f"✅ Системный промпт установлен: \"{SYSTEM_PROMPT}\"")


# --------------------------------------------------
#  5. Динамическая инициализация таблицы результатов
#  Автоматически создаёт колонки под любое количество вопросов
# --------------------------------------------------

def init_results_table():
    """
    Инициализирует или возвращает существующую таблицу результатов.
    Колонки динамически формируются под количество вопросов в TEST_CASES.
    """
    global results_df

    # Если таблица уже существует — не пересоздаём
    if 'results_df' in globals() and isinstance(results_df, pd.DataFrame) and not results_df.empty:
        print("ℹ️ Таблица результатов уже существует — оставляем как есть.")
        return results_df

    # Формируем названия колонок динамически — БЕЗ обрезки текста вопроса
    columns = ['Модель']

    for i, (_, _, question_text) in enumerate(TEST_CASES, 1):
        columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])

    columns.append('Общее время (сек)')
    columns.append('Итоговый вывод')  # ← для ручного заполнения после анализа

    # Создаём пустую таблицу
    results_df = pd.DataFrame(columns=columns)
    print(f"✅ Таблица результатов инициализирована под {len(TEST_CASES)} вопросов.")
    return results_df


# Инициализируем таблицу
results_df = init_results_table()


# --------------------------------------------------
#  6. УНИВЕРСАЛЬНАЯ функция добавления результатов в таблицу
#  Работает с ЛЮБЫМ количеством вопросов
# --------------------------------------------------

def add_to_results(model_name, answers, times, total_time):
    """
    УНИВЕРСАЛЬНАЯ функция: добавляет результаты тестирования модели в таблицу.
    Поддерживает ЛЮБОЕ количество вопросов.
    Итоговый вывод остаётся пустым — его заполняет пользователь вручную после анализа.

     Параметры:
        model_name : str
            Название модели (например, "Qwen/Qwen1.5-7B-Chat").
        answers : list[str]
            Список сгенерированных ответов — по одному на каждый вопрос.
        times : list[float]
            Список времён генерации — по одному на каждый ответ.
        total_time : float
            Общее время выполнения всех запросов.

     Пример вызова:
        result = test_model_universal(tokenizer, model)
        add_to_results("Saiga-7B", result["answers"], result["times"], result["total_time"])
    """
    global results_df

    # Проверка соответствия длины списков
    if len(answers) != len(times):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а времён {len(times)}")
    if len(answers) != len(TEST_QUESTIONS):
        raise ValueError(f"Несоответствие: ответов {len(answers)}, а вопросов {len(TEST_QUESTIONS)}")

    # Создаём новую строку результата
    new_row = {'Модель': model_name}

    # Заполняем данные по каждому вопросу
    for i in range(len(answers)):
        q_num = i + 1

        # Ищем точное имя колонки
        question_col = f"Вопрос {q_num}"
        if question_col not in results_df.columns:
            raise KeyError(f"Не найдена колонка для вопроса {q_num}: '{question_col}'")

        # Заполняем текст вопроса, ответ и время
        new_row[question_col] = TEST_QUESTIONS[i]
        new_row[f"Ответ {q_num}"] = answers[i]
        new_row[f"Время ответа {q_num} (сек)"] = round(times[i], 2)

    # Заполняем общее время и оставляем итог пустым
    new_row['Общее время (сек)'] = round(total_time, 2)
    new_row['Итоговый вывод'] = ""

    # Добавляем строку в DataFrame
    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)
    print(f"✅ Результаты для {model_name} добавлены в таблицу. Итоговый вывод — заполните вручную.")

# --------------------------------------------------
# 7. Функция добавления итогового вывода по модели
# --------------------------------------------------

def add_final_conclusion(model_name, conclusion):
    """
    Ручное добавление итогового анализа для указанной модели.
    Обновляет существующую строку в таблице results_df.

    Параметры:
        model_name : str
            Название модели, как оно было добавлено ранее (например, "Qwen/Qwen1.5-7B-Chat").
        conclusion : str
            Текст итогового вывода — анализ логики, ошибок, скорости и т.д.

    """
    global results_df

    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов не инициализирована.")
        return

    # Находим индекс строки с указанной моделью
    mask = results_df['Модель'] == model_name
    if not mask.any():
        print(f"❌ Модель '{model_name}' не найдена в таблице.")
        return

    # Обновляем значение в колонке "Итоговый вывод"
    results_df.loc[mask, 'Итоговый вывод'] = conclusion
    print(f"✅ Итоговый вывод для модели '{model_name}' успешно добавлен.")


# --------------------------------------------------
#  8. Функция вывода результатов (улучшенная для сравнения моделей)
# --------------------------------------------------

def show_results():
    """
    Выводит таблицу сравнения моделей с полными ответами и удобным форматированием.
    """
    if 'results_df' not in globals() or results_df.empty:
        print("⚠️ Таблица результатов пуста или не инициализирована.")
        return

    print("\n📊 СРАВНЕНИЕ МОДЕЛЕЙ НА ТЕСТОВЫХ ВОПРОСАХ\n")

    # Настройка pandas для полного отображения текста
    pd.set_option('display.max_colwidth', None)
    pd.set_option('display.width', None)
    pd.set_option('display.max_columns', None)

    # Определяем колонки для отображения
    cols_to_show = ['Модель']
    for i in range(1, len(TEST_QUESTIONS) + 1):
        cols_to_show.extend([f'Вопрос {i}', f'Ответ {i}'])
    cols_to_show.append('Общее время (сек)')
    cols_to_show.append('Итоговый вывод')

    # Проверка наличия колонок
    missing = [col for col in cols_to_show if col not in results_df.columns]
    if missing:
        print(f"⚠️ Отсутствуют колонки: {missing}. Возможно, таблица не инициализирована корректно.")
        display(results_df)
        return

    # Формируем сводную таблицу
    summary_df = results_df[cols_to_show].copy()

    # Стилизация
    styled = summary_df.style.set_properties(**{
        'text-align': 'left',
        'white-space': 'pre-wrap',
        'word-wrap': 'break-word',
        'vertical-align': 'top',
        'font-family': 'Consolas, monospace',
        'font-size': '13px'
    }).set_table_styles([
        {'selector': 'th', 'props': [('background-color', '#f0f0f0'), ('font-weight', 'bold')]},
        {'selector': 'td', 'props': [('border', '1px solid #ddd'), ('padding', '10px')]}
    ]).set_caption("📈 Сравнение локальных LLM на тестах логики и креативности")

    display(styled)



# ============================================
#  УНИВЕРСАЛЬНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ LLM
# ============================================

# --- ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ

def _infer_device(model):
    """Определяет, на каком устройстве (CPU/GPU) работает модель."""
    try:
        return next(model.parameters()).device
    except StopIteration:
        return torch.device("cuda") if torch.cuda.is_available() else torch.device("cpu")

def _ensure_padding_ids(tokenizer, model):
    """Гарантирует, что pad_token_id задан — иначе используем eos_token_id."""
    if tokenizer.pad_token_id is None and tokenizer.eos_token_id is not None:
        tokenizer.pad_token_id = tokenizer.eos_token_id

def _maybe_inject_system(messages, system_prompt):
    """Добавляет системный промпт в начало, если его нет и он задан."""
    if not system_prompt:
        return list(messages)
    if any(m.get("role") == "system" for m in messages):
        return list(messages)
    return [{"role": "system", "content": system_prompt}] + list(messages)

def _build_fallback_prompt(messages, system_prompt=None):
    """Создаёт простой текстовый промпт, если у токенизатора нет chat_template."""
    parts = []
    if system_prompt:
        parts.append(f"system\n{system_prompt}\n")
    for m in messages:
        parts.append(f"{m.get('role', 'user')}\n{m.get('content', '')}\n")
    parts.append("assistant\n")
    return "".join(parts)


# --- ОСНОВНАЯ ФУНКЦИЯ ТЕСТИРОВАНИЯ ---

def test_model_universal(
    tokenizer,
    model,
    device=None,
    *,
    system_prompt=None,
    generation_config=None,
    max_new_tokens=200,
    temperature=0.7,
    top_p=0.9,
    show_prompts=True
):
    """
    Тестирует модель на всех вопросах из TEST_CASES.
    Возвращает структурированный словарь с ответами, временами и общим временем.
    Подходит для любых моделей и любого числа вопросов.
    """

    # Подготавливаем окружение
    _ensure_padding_ids(tokenizer, model)
    if device is None:
        device = _infer_device(model)

    answers, times = [], []
    total_time = 0.0

    # Если нет вопросов — возвращаем пустой результат
    if not TEST_CASES:
        print("⚠️  Нет тестовых вопросов — возвращаем пустой результат.")
        return {"answers": [], "times": [], "total_time": 0.0}

    # Обрабатываем каждый вопрос
    for question_num, messages, question_text in TEST_CASES:
        print(f"🔹 Вопрос {question_num}: {question_text[:80]}...")

        # Формируем сообщения с системным промптом (если нужен)
        msgs = _maybe_inject_system(messages, system_prompt)

        # Генерируем промпт — через шаблон или fallback
        try:
            prompt = tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
        except (ValueError, AttributeError, NotImplementedError):
            print("⚠️  Без chat_template — используем fallback-формат.")
            prompt = _build_fallback_prompt(msgs, system_prompt=system_prompt)

        if show_prompts:
            print(f"📝 Промпт:\n{prompt}\n{'='*60}")

        # Токенизируем и переносим на устройство
        inputs = tokenizer(prompt, return_tensors="pt", add_special_tokens=False)
        inputs = {k: v.to(device) for k, v in inputs.items()}

        # Генерируем ответ
        start_time = time.time()
        with torch.no_grad():
            if generation_config is not None:
                # Используем переданный конфиг генерации
                gc = generation_config
                if getattr(gc, "pad_token_id", None) is None and tokenizer.pad_token_id is not None:
                    gc = gc.clone()
                    gc.pad_token_id = tokenizer.pad_token_id
                outputs = model.generate(**inputs, generation_config=gc)
            else:
                # Используем стандартные параметры
                outputs = model.generate(
                    **inputs,
                    max_new_tokens=max_new_tokens,
                    do_sample=True,
                    temperature=temperature,
                    top_p=top_p,
                    pad_token_id=tokenizer.pad_token_id,
                    eos_token_id=getattr(model.config, "eos_token_id", None)
                )

        gen_time = time.time() - start_time
        total_time += gen_time

        # Декодируем только сгенерированную часть (после входного промпта)
        out_ids = outputs[0]
        in_len = inputs["input_ids"].shape[1]
        gen_ids = out_ids[in_len:]
        answer = tokenizer.decode(gen_ids, skip_special_tokens=True).strip()

        # Выводим результат
        print(f"✅ Ответ {question_num} за {gen_time:.2f} сек")
        print(f"💬 {answer}\n")

        answers.append(answer)
        times.append(gen_time)

    # Возвращаем всё в структурированном виде — для любого числа вопросов
    return {
        "answers": answers,
        "times": times,
        "total_time": total_time
    }


# --------------------------------------------------
#  1. Путь по умолчанию для сохранения результатов на Google Drive
#  Легко переопределить при вызове функций
# --------------------------------------------------

DEFAULT_SAVE_PATH = "/content/drive/MyDrive/llm_benchmark/llm_comparison_results.csv"


# --------------------------------------------------
#  2. Умная загрузка результатов с адаптацией структуры
#  Загружает CSV и автоматически приводит к актуальной схеме вопросов
#  Если файл отсутствует — создаёт новую таблицу
#  Гарантирует совместимость с add_to_results()
# --------------------------------------------------

def load_or_init_results(test_questions=TEST_QUESTIONS, filepath=DEFAULT_SAVE_PATH):
    """
    Загружает сохранённые результаты из CSV и адаптирует структуру под текущие вопросы.
    Если файла нет — создаёт новую таблицу.
    Возвращает DataFrame `results_df`.

    Параметры:
        test_questions : list — список вопросов (по умолчанию TEST_QUESTIONS)
        filepath : str — путь к файлу CSV

    Использование:
        results_df = load_or_init_results()
    """
    # Формируем ожидаемую структуру колонок на основе текущих вопросов
    expected_columns = ['Модель']
    for i in range(1, len(test_questions) + 1):
        expected_columns.extend([
            f'Вопрос {i}',
            f'Ответ {i}',
            f'Время ответа {i} (сек)'
        ])
    expected_columns.extend(['Общее время (сек)', 'Итоговый вывод'])

    #  Проверяем наличие файла
    if os.path.exists(filepath):
        try:
            #  Загружаем данные
            loaded_df = pd.read_csv(filepath)
            print(f"📂 Загружены предыдущие результаты из: {filepath}")

            #  Проверяем, совпадает ли структура
            if set(loaded_df.columns) == set(expected_columns):
                results_df = loaded_df.copy()
                print(f"✅ Структура актуальна. В таблице {len(results_df)} моделей.")
            else:
                # Адаптируем старую таблицу под новую структуру
                print("⚠️ Структура устарела — создаём обновлённую версию...")
                results_df = pd.DataFrame(columns=expected_columns)

                #  Переносим совместимые данные
                for _, row in loaded_df.iterrows():
                    new_row = {}
                    # Общие поля
                    for col in ['Модель', 'Общее время (сек)', 'Итоговый вывод']:
                        new_row[col] = row.get(col, "")
                    # Поля по каждому вопросу
                    for i in range(1, len(test_questions) + 1):
                        for ft in ['Вопрос', 'Ответ']:
                            new_row[f"{ft} {i}"] = row.get(f"{ft} {i}", "")
                        new_row[f"Время ответа {i} (сек)"] = row.get(f"Время ответа {i} (сек)", 0.0)
                    # Добавляем строку
                    results_df = pd.concat([results_df, pd.DataFrame([new_row])], ignore_index=True)

                print(f"✅ Создана обновлённая таблица с {len(results_df)} перенесёнными записями.")

        except Exception as e:
            # При ошибке — создаём новую таблицу
            print(f"❌ Ошибка загрузки: {e}. Создаём новую таблицу.")
            results_df = pd.DataFrame(columns=expected_columns)
    else:
        # Файл не найден — инициализируем с нуля
        print("🆕 Файл не найден — создаём новую таблицу.")
        results_df = pd.DataFrame(columns=expected_columns)
        print(f"✅ Таблица инициализирована под {len(test_questions)} вопросов.")

    # Сбрасываем индекс для чистоты
    return results_df.reset_index(drop=True)


# --------------------------------------------------
# 3. Надёжное сохранение чекпоинта на Google Drive
#  Создаёт папку, если её нет
#  Кодировка utf-8-sig для корректного открытия в Excel
#  Выводит подтверждение — можно перезапускать среду
# --------------------------------------------------

def save_results_checkpoint(df, filepath=DEFAULT_SAVE_PATH):
    """
    Сохраняет DataFrame в CSV на Google Drive.

    Параметры:
        df : pd.DataFrame — таблица для сохранения
        filepath : str — путь к файлу CSV

     Использование:
        save_results_checkpoint(results_df)
    """
    #  Создаём директорию, если не существует
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    #  Сохраняем с корректной кодировкой
    df.to_csv(filepath, index=False, encoding='utf-8-sig')
    print(f"💾 Чекпоинт сохранён: {filepath}")
    print("✅ Теперь можно перезапустить среду.")


# Авторизация в Hugging Face
hf_token = userdata.get('HF_TOKEN')
login(token=hf_token)
print("✅ Авторизованы в Hugging Face")
````
